```python
#@title 🧬 Protein Sequence Input { display-mode: "form" }

input_mode = "Manual input only"

manual_sequence_1 = "MALWMRLLPLLALLALWGPDPAAAFVNQHLCGSHLVEALYLVCGERGFFYTPKTRREAEDLQVGQVELGGGPGAGSLQPLALEGSLQKRGIVEQCCTSICSLYQLENYCN" #@param {type:"string"}
manual_sequence_2 = "MGAKAPRGPKVAQWAMETAVIGVVVVLFVVTVAITCVLCCFSCDSRAQDPQGGPGRSFTVATFRQEASLFTGPVRHAQPVPSAQDFWTFM" #@param {type:"string"}
manual_sequence_3 = "" #@param {type:"string"}
manual_sequence_4 = "" #@param {type:"string"}
manual_sequence_5 = "" #@param {type:"string"}
manual_sequence_6 = "" #@param {type:"string"}
manual_sequence_7 = "" #@param {type:"string"}
manual_sequence_8 = "" #@param {type:"string"}
manual_sequence_9 = "" #@param {type:"string"}
manual_sequence_10 = "" #@param {type:"string"}
manual_sequence_11 = "" #@param {type:"string"}
manual_sequence_12 = "" #@param {type:"string"}
manual_sequence_13 = "" #@param {type:"string"}
manual_sequence_14 = "" #@param {type:"string"}
manual_sequence_15 = "" #@param {type:"string"}
manual_sequence_16 = "" #@param {type:"string"}
manual_sequence_17 = "" #@param {type:"string"}
manual_sequence_18 = "" #@param {type:"string"}
manual_sequence_19 = "" #@param {type:"string"}
manual_sequence_20 = "" #@param {type:"string"}
manual_sequence_21 = "" #@param {type:"string"}
manual_sequence_22 = "" #@param {type:"string"}
manual_sequence_23 = "" #@param {type:"string"}
manual_sequence_24 = "" #@param {type:"string"}
manual_sequence_25 = "" #@param {type:"string"}

import re
from pathlib import Path
from html import escape
from IPython.display import HTML, display
import pandas as pd

# ============================================================
# Project Directory
# ============================================================

if Path("/content/drive/MyDrive").exists():
    PROJECT_DIR = Path("/content/drive/MyDrive/FoldColabFlow_Project")
else:
    PROJECT_DIR = Path("/content/FoldColabFlow_Project")

INPUT_DIR = PROJECT_DIR / "inputs"
INPUT_DIR.mkdir(parents=True, exist_ok=True)

# ============================================================
# Input Quality Settings
# ============================================================

# Conservative pipeline alphabet:
# 20 standard amino acid residues + X as unknown residue.
STANDARD_AA = set("ACDEFGHIKLMNPQRSTVWY")
UNKNOWN_AA = set("X")

# U, O, B, Z, and J are not treated as biologically invalid.
# They are retained but flagged because they are outside the conservative 20 AA + X alphabet.
UNSUPPORTED_WARNING_AA = set("UOBJZ")

PIPELINE_ACCEPTED_AA = STANDARD_AA | UNKNOWN_AA
WARNING_RETAINED_AA = PIPELINE_ACCEPTED_AA | UNSUPPORTED_WARNING_AA

MIN_SEQUENCE_LENGTH_WARNING = 20
LONG_SEQUENCE_LENGTH_WARNING = 1500
HIGH_X_PERCENT_WARNING = 5.0
LOW_COMPLEXITY_PERCENT_WARNING = 60.0
NUCLEOTIDE_LIKE_MIN_LENGTH_WARNING = 30

MAX_MANUAL_SEQUENCES = 25

sequences = []
seq_names = []
selected_files = []

loaded_rows = []
warning_rows = []

sequence_quality_warnings = []
used_sequence_names = {}

# ============================================================
# Helper Functions
# ============================================================

def clean_sequence(seq):
    """
    Remove spaces and line breaks, then convert to uppercase.
    """
    seq = re.sub(r"\s+", "", str(seq))
    seq = seq.upper()
    return seq


def safe_sequence_name(name, fallback):
    """
    Convert sequence name into a safe file/report name.
    """
    name = str(name).strip()
    name = re.sub(r"[^A-Za-z0-9_.-]+", "_", name)
    name = name.strip("_")
    return name if name else fallback


def make_unique_sequence_name(base_name):
    """
    Make sequence names unique to avoid overwriting outputs in later stages.
    """
    base_name = safe_sequence_name(base_name, "Seq")

    if base_name not in used_sequence_names:
        used_sequence_names[base_name] = 1
        return base_name, False

    used_sequence_names[base_name] += 1
    new_name = f"{base_name}_{used_sequence_names[base_name]}"

    return new_name, True


def add_quality_warning(protein_name, warning_type, details, recommendation):
    """
    Store biological/input quality warnings for display and final report.
    """
    sequence_quality_warnings.append({
        "Protein Name": protein_name,
        "Warning Type": warning_type,
        "Details": details,
        "Recommendation": recommendation
    })


def detect_quality_warnings(final_name, seq, renamed_from_duplicate_name=False, duplicate_of=None):
    """
    Detect biological/input quality issues as warning-level findings.
    Non-empty sequences are retained for downstream analysis but clearly flagged when needed.
    """
    local_warnings = []
    seq_len = len(seq)

    # 1. Unexpected symbols outside the retained alphabet
    unexpected_chars = sorted(set(seq) - WARNING_RETAINED_AA)

    if unexpected_chars:
        local_warnings.append((
            "Unexpected sequence symbols",
            f"Contains unexpected symbols: {', '.join(unexpected_chars)}.",
            "The sequence was retained, but these symbols may not be compatible with downstream tools. Review the input before ColabFold-based prediction or downstream analysis."
        ))

    # 2. Unsupported or ambiguous residue symbols
    unsupported_symbols = sorted(set(seq) & UNSUPPORTED_WARNING_AA)

    if unsupported_symbols:
        symbol_notes = {
            "U": "selenocysteine",
            "O": "pyrrolysine",
            "B": "aspartic acid/asparagine ambiguity",
            "Z": "glutamic acid/glutamine ambiguity",
            "J": "leucine/isoleucine ambiguity"
        }

        details_list = [
            f"{symbol} ({symbol_notes.get(symbol, 'ambiguous residue symbol')})"
            for symbol in unsupported_symbols
        ]

        local_warnings.append((
            "Unsupported or ambiguous residue symbols",
            "Contains " + ", ".join(details_list) + ".",
            "These symbols are not treated as biologically invalid, but they are outside the conservative 20 amino acid + X alphabet and may require manual review before downstream analysis."
        ))

    # 3. Very short sequence
    if seq_len < MIN_SEQUENCE_LENGTH_WARNING:
        local_warnings.append((
            "Very short sequence",
            f"Sequence length is {seq_len}.",
            "This is a pipeline-defined operational warning. Very short sequences may provide limited information for comparative alignment or structure prediction."
        ))

    # 4. Very long sequence
    if seq_len > LONG_SEQUENCE_LENGTH_WARNING:
        local_warnings.append((
            "Very long sequence",
            f"Sequence length is {seq_len}.",
            "This is a conservative operational warning. Long protein inputs may increase memory use and runtime in ColabFold-based structure prediction and visualization."
        ))

    # 5. Unknown residue X
    x_count = seq.count("X")

    if x_count > 0:
        x_percent = (x_count / seq_len) * 100 if seq_len else 0

        if x_percent > HIGH_X_PERCENT_WARNING:
            local_warnings.append((
                "High unknown-residue content",
                f"Contains {x_count} X residues ({x_percent:.2f}%).",
                "This is a pipeline-defined warning threshold. High X content may reduce the reliability of alignment, physicochemical analysis, and structure prediction."
            ))
        else:
            local_warnings.append((
                "Unknown residue detected",
                f"Contains {x_count} X residues ({x_percent:.2f}%).",
                "X is accepted as an unknown amino acid, but downstream biological interpretation should be cautious."
            ))

    # 6. Duplicate sequence content
    if duplicate_of is not None:
        local_warnings.append((
            "Duplicate sequence",
            f"This sequence is identical to {duplicate_of}.",
            "Duplicate proteins may create redundant pairwise comparisons and repeated outputs."
        ))

    # 7. Duplicate protein name renamed
    if renamed_from_duplicate_name:
        local_warnings.append((
            "Duplicate protein name",
            "The original sequence name was already used and was automatically renamed.",
            "Automatic renaming was applied to preserve unique identifiers and avoid output file overwriting."
        ))

    # 8. Possible low-complexity composition / residue dominance
    if seq_len > 0:
        residue_counts = {}

        for residue in seq:
            residue_counts[residue] = residue_counts.get(residue, 0) + 1

        most_common_residue = max(residue_counts, key=residue_counts.get)
        most_common_percent = (residue_counts[most_common_residue] / seq_len) * 100

        if most_common_percent >= LOW_COMPLEXITY_PERCENT_WARNING:
            local_warnings.append((
                "Possible low-complexity composition",
                f"Residue {most_common_residue} represents {most_common_percent:.2f}% of the sequence.",
                "This residue-dominance rule is an operational warning for strong compositional bias and does not replace formal low-complexity-region detection algorithms."
            ))

    # 9. Possible nucleotide input
    nucleotide_letters = set("ACGTUN")

    if seq_len >= NUCLEOTIDE_LIKE_MIN_LENGTH_WARNING and set(seq).issubset(nucleotide_letters):
        local_warnings.append((
            "Possible nucleotide input",
            "The sequence is at least 30 characters long and uses only A, C, G, T, U and N.",
            "Because these letters can overlap with valid one-letter amino acid codes, the sequence was retained but should be manually reviewed as a possible DNA/RNA input."
        ))

    for warning_type, details, recommendation in local_warnings:
        add_quality_warning(final_name, warning_type, details, recommendation)

    return local_warnings


def add_sequence(source_name, record_name, seq, max_allowed):
    """
    Add one sequence to the global sequence list.
    This implementation applies warning-based screening and does not reject non-empty sequences.
    """
    if max_allowed and len(sequences) >= max_allowed:
        warning_rows.append(f"""
        <tr>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">{escape(source_name)}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">{escape(record_name)}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">Skipped because the maximum sequence limit was reached.</td>
        </tr>
        """)
        return

    if not seq:
        warning_rows.append(f"""
        <tr>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">{escape(source_name)}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">{escape(record_name)}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fecaca;">Empty sequence skipped.</td>
        </tr>
        """)
        return

    seq_index = len(sequences) + 1
    base_name = safe_sequence_name(record_name, f"Seq_{seq_index}")
    final_name, renamed_from_duplicate_name = make_unique_sequence_name(base_name)

    duplicate_of = None

    if seq in sequences:
        duplicate_index = sequences.index(seq)
        duplicate_of = seq_names[duplicate_index]

    local_warnings = detect_quality_warnings(
        final_name=final_name,
        seq=seq,
        renamed_from_duplicate_name=renamed_from_duplicate_name,
        duplicate_of=duplicate_of
    )

    sequences.append(seq)
    seq_names.append(final_name)

    if local_warnings:
        status_text = '<span style="color:#b91c1c; font-weight:bold;">Loaded with warning</span>'
        row_bg = "#fff1f2"
    else:
        status_text = '<span style="color:#15803d; font-weight:bold;">Loaded</span>'
        row_bg = "#ffffff"

    loaded_rows.append(f"""
    <tr style="background-color:{row_bg};">
        <td style="text-align:center; padding:8px; border:1px solid #cbd5e1;">{escape(source_name)}</td>
        <td style="text-align:center; padding:8px; border:1px solid #cbd5e1;"><b>{escape(final_name)}</b></td>
        <td style="text-align:center; padding:8px; border:1px solid #cbd5e1;">{len(seq)}</td>
        <td style="text-align:center; padding:8px; border:1px solid #cbd5e1;">{status_text}</td>
    </tr>
    """)

# ============================================================
# Input Loading
# ============================================================

manual_inputs = [
    manual_sequence_1,
    manual_sequence_2,
    manual_sequence_3,
    manual_sequence_4,
    manual_sequence_5,
    manual_sequence_6,
    manual_sequence_7,
    manual_sequence_8,
    manual_sequence_9,
    manual_sequence_10,
    manual_sequence_11,
    manual_sequence_12,
    manual_sequence_13,
    manual_sequence_14,
    manual_sequence_15,
    manual_sequence_16,
    manual_sequence_17,
    manual_sequence_18,
    manual_sequence_19,
    manual_sequence_20,
    manual_sequence_21,
    manual_sequence_22,
    manual_sequence_23,
    manual_sequence_24,
    manual_sequence_25
]

for i, raw_text in enumerate(manual_inputs, start=1):
    raw_text = str(raw_text).strip()

    if not raw_text:
        continue

    seq = clean_sequence(raw_text)

    add_sequence(
        source_name="Manual input",
        record_name=f"Seq_{i}",
        seq=seq,
        max_allowed=MAX_MANUAL_SEQUENCES
    )

# ============================================================
# Create warning dataframe for downstream report
# ============================================================

sequence_quality_warning_df = pd.DataFrame(sequence_quality_warnings)

# ============================================================
# Styled Output
# ============================================================

loaded_table = "".join(loaded_rows) if loaded_rows else """
<tr>
    <td colspan="4" style="color:#b91c1c; font-weight:bold; text-align:center; padding:8px;">
        No protein sequences were loaded.
    </td>
</tr>
"""

basic_warning_section = ""

if warning_rows:
    basic_warning_section = f"""
    <div style="
        background-color:#fee2e2;
        border:2px solid #f87171;
        border-radius:12px;
        padding:14px;
        margin-top:16px;
    ">
        <h3 style="color:#991b1b; margin-top:0;">⚠️ Input Warnings</h3>

        <table style="
            width:100%;
            border-collapse:collapse;
            font-size:14px;
            table-layout:fixed;
            word-wrap:break-word;
            background-color:#ffffff;
        ">
            <tr style="background-color:#fecaca;">
                <th style="text-align:center; padding:8px; border:1px solid #f87171;">Source</th>
                <th style="text-align:center; padding:8px; border:1px solid #f87171;">Record</th>
                <th style="text-align:center; padding:8px; border:1px solid #f87171;">Message</th>
            </tr>
            {"".join(warning_rows)}
        </table>
    </div>
    """

quality_warning_section = ""

if sequence_quality_warnings:
    quality_warning_rows = ""

    for item in sequence_quality_warnings:
        quality_warning_rows += f"""
        <tr>
            <td style="text-align:center; padding:8px; border:1px solid #fca5a5;"><b>{escape(item["Protein Name"])}</b></td>
            <td style="text-align:center; padding:8px; border:1px solid #fca5a5;">{escape(item["Warning Type"])}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fca5a5;">{escape(item["Details"])}</td>
            <td style="text-align:center; padding:8px; border:1px solid #fca5a5;">{escape(item["Recommendation"])}</td>
        </tr>
        """

    quality_warning_section = f"""
    <div style="
        background-color:#fee2e2;
        border:2px solid #ef4444;
        border-radius:12px;
        padding:14px;
        margin-top:16px;
    ">
        <h3 style="color:#991b1b; margin-top:0;">🚨 Biological / Input Quality Warnings</h3>

        <p style="color:#7f1d1d; line-height:1.8; margin-top:0;">
            These sequences were loaded, but they contain warning-level issues that may affect alignment,
            physicochemical analysis, structure prediction, or downstream interpretation.
        </p>

        <table style="
            width:100%;
            border-collapse:collapse;
            font-size:14px;
            table-layout:fixed;
            word-wrap:break-word;
            background-color:#ffffff;
        ">
            <tr style="background-color:#fecaca;">
                <th style="text-align:center; padding:8px; border:1px solid #ef4444;">Protein Name</th>
                <th style="text-align:center; padding:8px; border:1px solid #ef4444;">Warning Type</th>
                <th style="text-align:center; padding:8px; border:1px solid #ef4444;">Details</th>
                <th style="text-align:center; padding:8px; border:1px solid #ef4444;">Recommendation</th>
            </tr>
            {quality_warning_rows}
        </table>
    </div>
    """
else:
    quality_warning_section = """
    <div style="
        background-color:#dcfce7;
        border:1px solid #86efac;
        border-radius:12px;
        padding:14px;
        margin-top:16px;
    ">
        <p style="margin:0; color:#166534; font-size:15px;">
            ✅ No biological/input quality warnings detected for loaded sequences.
        </p>
    </div>
    """

quality_status = "Requires attention" if sequence_quality_warnings or warning_rows else "Passed"
quality_status_color = "#b91c1c" if sequence_quality_warnings or warning_rows else "#15803d"

display(HTML(f"""
<div style="
    background-color:#e5e7eb;
    border:2px solid #9ca3af;
    border-radius:16px;
    padding:22px;
    font-family:Arial, sans-serif;
    color:#1f2937;
    margin:16px 0;
    overflow:visible;
    max-height:none;
">

    <h2 style="color:#1d4ed8; margin-top:0; text-align:center; margin-bottom:6px;">
        🧬 Protein Sequence Input
    </h2>

    <p style="
        text-align:center;
        color:#4b5563;
        font-size:14px;
        margin-top:0;
        margin-bottom:18px;
        line-height:1.7;
    ">
        Enter up to 25 protein sequences manually.
    </p>

    <div style="
        background-color:#f8fafc;
        border:1px solid #cbd5e1;
        border-radius:12px;
        padding:14px;
        margin-bottom:16px;
    ">
        <h3 style="color:#374151; margin-top:0;">📍 CURRENT STAGE</h3>
        <p style="margin-bottom:0; font-size:16px;">
            Protein Sequence Input
        </p>
    </div>

    <div style="
        background-color:#f8fafc;
        border:1px solid #cbd5e1;
        border-radius:12px;
        padding:14px;
        margin-bottom:16px;
    ">
        <h3 style="color:#374151; margin-top:0;">✨ INPUT NOTES</h3>

        <p style="font-size:15px; line-height:1.9; margin-bottom:0;">
            ✨ Manual input supports up to <b>25 protein sequences</b><br>
            💡 Leave unused manual sequence boxes empty<br>
            💡 Spaces and line breaks will be removed automatically<br>
            💡 Sequences will be converted to uppercase before analysis<br>
            💡 Conservative alphabet: 20 standard amino acids + X<br>
            💡 U, O, B, Z, and J are retained but flagged for manual review<br>
            🚨 Sequences with suspicious quality issues will be loaded but clearly flagged in red
        </p>
    </div>

    <div style="
        background-color:#f8fafc;
        border:1px solid #cbd5e1;
        border-radius:12px;
        padding:14px;
        margin-bottom:16px;
    ">
        <p style="margin:0; line-height:1.8;">
            <b>Input mode:</b> {escape(input_mode)}<br>
            <b>Sequences loaded:</b> <span style="color:#15803d; font-weight:bold;">{len(sequences)}</span><br>
            <b>Quality status:</b> <span style="color:{quality_status_color}; font-weight:bold;">{quality_status}</span><br>
            <b>Project input directory:</b> <code>{escape(str(INPUT_DIR))}</code>
        </p>
    </div>

    <table style="
        width:100%;
        border-collapse:collapse;
        font-size:14px;
        table-layout:fixed;
        word-wrap:break-word;
        background-color:#ffffff;
    ">
        <tr style="background-color:#dbeafe;">
            <th style="text-align:center; padding:8px; border:1px solid #94a3b8;">Source</th>
            <th style="text-align:center; padding:8px; border:1px solid #94a3b8;">Sequence Name</th>
            <th style="text-align:center; padding:8px; border:1px solid #94a3b8;">Length</th>
            <th style="text-align:center; padding:8px; border:1px solid #94a3b8;">Status</th>
        </tr>
        {loaded_table}
    </table>

    {quality_warning_section}
    {basic_warning_section}

    <div style="
        background-color:#dcfce7;
        border:1px solid #86efac;
        border-radius:12px;
        padding:14px;
        margin-top:16px;
    ">
        <p style="margin:0; color:#166534; font-size:16px;">
            ➡️ <b>Next stage:</b> Sequence Alignment & Similarity Analysis
        </p>
    </div>

</div>

<script>
  google.colab.output.setIframeHeight(0, true, {{maxHeight: 100000}});
</script>
"""))

# ============================================================
# Optional State Saving
# ============================================================

if "save_state" in globals():
    save_state(
        "01_protein_sequence_input",
        sequences=sequences,
        seq_names=seq_names,
        input_mode=input_mode,
        input_dir=str(INPUT_DIR),
        sequence_quality_warnings=sequence_quality_warnings,
        sequence_quality_warning_df=sequence_quality_warning_df
    )
```

```python
#@title 🔬 Sequence Alignment & Similarity Analysis { display-mode: "form" }

alignment_analysis_mode = "Both" #@param ["All-vs-All Pairwise + Reference Summary", "Multiple Sequence Alignment with MAFFT", "Both"]

reference_protein_name = "" #@param {type:"string"}
use_custom_reference_sequence = False #@param {type:"boolean"}
custom_reference_sequence = "" #@param {type:"string"}
export_alignment_html_report = True #@param {type:"boolean"}

# ============================================================
# Internal settings
# ============================================================

MAFFT_STRATEGY      = "auto"
MSA_BLOCK_WIDTH     = 80
TOP_PAIRWISE_ROWS_TO_SHOW      = 7
IDENTITY_MATRIX_PREVIEW_LIMIT  = 6
MAX_DIRECT_DOWNLOAD_MB         = 30
COUNT_X_AS_IDENTICAL          = True   # retained for backward-compatible string identity; reported in settings

# ============================================================
# Install required packages
# ============================================================

import sys
import subprocess
import shutil
from pathlib import Path

subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q", "biopython", "pandas"],
    check=True
)

need_mafft = alignment_analysis_mode in [
    "Multiple Sequence Alignment with MAFFT", "Both"
]

if need_mafft and shutil.which("mafft") is None:
    subprocess.run(
        "apt-get -qq update && apt-get -qq install -y mafft",
        shell=True, check=True,
        stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL
    )

# ============================================================
# Imports
# ============================================================

import re
import itertools
import base64
import zipfile
import platform
from collections import Counter
from html import escape
from datetime import datetime

import pandas as pd
from IPython.display import HTML, display

import Bio
from Bio.Align import PairwiseAligner, substitution_matrices
from Bio import SeqIO, AlignIO
from Bio.SeqRecord import SeqRecord
from Bio.Seq import Seq

# ============================================================
# Global style
# ============================================================

STYLE_BLOCK = """
<style>
.fi-container {
    background-color:#e5e7eb;
    border:2px solid #9ca3af;
    border-radius:16px;
    padding:22px;
    font-family:Arial, sans-serif;
    color:#1f2937;
    margin:16px 0;
    overflow:visible;
    max-height:none;
}
.fi-title {
    text-align:center;
    color:#1d4ed8;
    margin-top:0;
    margin-bottom:18px;
}
.fi-box {
    background-color:#f8fafc;
    border:1px solid #cbd5e1;
    border-radius:12px;
    padding:14px;
    margin-bottom:16px;
}
.fi-box h3, .fi-box h4 { color:#374151; margin-top:0; }
.fi-next {
    background-color:#dcfce7;
    border:1px solid #86efac;
    border-radius:12px;
    padding:14px;
    margin-top:16px;
    color:#166534;
}
.fi-table {
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    margin:10px 0 12px 0;
    background-color:#ffffff;
}
.fi-table th {
    background-color:#dbeafe;
    color:#111827;
    text-align:center !important;
    padding:10px;
    border:1px solid #94a3b8;
    font-weight:bold;
}
.fi-table td {
    background-color:#ffffff;
    color:#1f2937;
    text-align:center !important;
    padding:10px;
    border:1px solid #cbd5e1;
    vertical-align:middle;
}
.fi-table tr:nth-child(even) td { background-color:#f8fafc; }
.fi-table tr:nth-child(odd)  td { background-color:#ffffff; }
.fi-small { font-size:13px; color:#4b5563; line-height:1.7; }
.fi-download {
    display:inline-block;
    background-color:#eff6ff;
    border:1px solid #93c5fd;
    color:#1d4ed8;
    padding:8px 12px;
    border-radius:10px;
    text-decoration:none;
    font-weight:bold;
    margin:4px 6px 4px 0;
    font-size:13px;
}
.fi-download:hover { background-color:#dbeafe; }
.fi-download-note {
    display:inline-block;
    background-color:#f1f5f9;
    border:1px solid #cbd5e1;
    color:#475569;
    padding:8px 12px;
    border-radius:10px;
    margin:4px 6px 4px 0;
    font-size:13px;
}
.fi-stat-grid {
    display:grid;
    grid-template-columns:repeat(auto-fit, minmax(170px, 1fr));
    gap:12px;
    margin-top:10px;
}
.fi-stat-card {
    background-color:#ffffff;
    border:1px solid #cbd5e1;
    border-radius:12px;
    padding:12px;
    text-align:center;
}
.fi-stat-label { font-size:13px; color:#64748b; margin-bottom:6px; }
.fi-stat-value { font-size:18px; font-weight:bold; color:#1f2937; }

/* ── CLUSTAL viewer ── */
.aln-wrap {
    background:#1e1e2e;
    border-radius:10px;
    padding:14px 16px;
    overflow-x:auto;
    margin-top:10px;
}
.aln-block {
    margin-bottom:18px;
}
.aln-block-title {
    color:#a6adc8;
    font-size:11px;
    font-family:monospace;
    margin-bottom:6px;
    letter-spacing:.05em;
}
.aln-row {
    display:flex;
    align-items:baseline;
    line-height:1.55;
}
.aln-label {
    color:#cdd6f4;
    font-family:monospace;
    font-size:13px;
    min-width:130px;
    flex-shrink:0;
    overflow:hidden;
    text-overflow:ellipsis;
    white-space:nowrap;
    padding-right:8px;
}
.aln-seq {
    font-family:monospace;
    font-size:13px;
    white-space:pre;
    letter-spacing:.05em;
}
.aln-match {
    font-family:monospace;
    font-size:13px;
    white-space:pre;
    color:#6c7086;
    letter-spacing:.05em;
    min-width:130px;
    padding-left:138px;
}
/* amino-acid colours (ClustalX-like scheme) */
.aa-hydrophobic { color:#f38ba8; }   /* A V I L M F W P  – red   */
.aa-positive    { color:#89b4fa; }   /* K R H             – blue  */
.aa-negative    { color:#f9e2af; }   /* D E               – yellow*/
.aa-polar       { color:#a6e3a1; }   /* S T N Q           – green */
.aa-cysteine    { color:#fab387; }   /* C                 – orange*/
.aa-glycine     { color:#cba6f7; }   /* G                 – purple*/
.aa-proline     { color:#94e2d5; }   /* P                 – cyan  */
.aa-aromatic    { color:#f5c2e7; }   /* F Y W             – pink  */
.aa-gap         { color:#45475a; }   /* -                 – grey  */
.aa-match-id    { color:#a6e3a1; font-weight:bold; }   /* | identical  */
.aa-match-cons  { color:#f9e2af; }                     /* + conserved  */
.aa-match-none  { color:#45475a; }                     /* . mismatch   */

.aln-legend {
    display:flex;
    flex-wrap:wrap;
    gap:10px;
    margin-top:8px;
    font-size:12px;
    font-family:monospace;
}
.aln-legend-item { display:flex; align-items:center; gap:4px; color:#cdd6f4; }
.aln-legend-dot  { width:12px; height:12px; border-radius:3px; display:inline-block; }
</style>
"""

RESIZE_SCRIPT = """
<script>
  try { google.colab.output.setIframeHeight(0, true, {maxHeight: 1000000}); } catch(e) {}
</script>
"""

# ============================================================
# Standalone HTML Report Styling (Green + Grey theme)
# ============================================================

ALIGNMENT_REPORT_CSS = """
<style>
:root {
    --accent:#74c69d;
    --accent-dark:#2d6a4f;
    --accent-soft:#eaf7ee;
    --accent-line:#b7e4c7;
    --slate:#111827;
    --muted:#64748b;
    --border:#d1d5db;
    --bg:#f3f4f6;
    --card:#ffffff;
    --card-soft:#f8fafc;
    --warning:#fef3c7;
    --danger:#fee2e2;
}
* { box-sizing:border-box; }
html { scroll-behavior:smooth; }
body {
    margin:0; padding:0;
    font-family:Inter, ui-sans-serif, system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", Arial, sans-serif;
    background:
        radial-gradient(circle at top left, rgba(22,163,74,0.12), transparent 32%),
        linear-gradient(180deg, #f8fafc 0%, var(--bg) 100%);
    color:var(--slate);
    line-height:1.65;
}
header {
    background:linear-gradient(135deg, #14532d 0%, #15803d 48%, #4b5563 100%);
    color:white;
    padding:38px 46px;
    border-bottom:5px solid var(--accent-line);
    box-shadow:0 14px 34px rgba(15,23,42,0.16);
}
header h1 { margin:0 0 8px 0; font-size:31px; line-height:1.2; font-weight:850; letter-spacing:-0.02em; }
header p { margin:0; opacity:0.93; font-size:14px; }
.layout {
    display:grid;
    grid-template-columns:280px minmax(0, 1fr);
    gap:24px;
    padding:26px;
    max-width:1720px;
    margin:0 auto;
}
.sidebar {
    position:sticky; top:18px; align-self:start;
    background:rgba(255,255,255,0.94);
    border:1px solid var(--border);
    border-radius:18px;
    padding:18px;
    box-shadow:0 10px 26px rgba(15,23,42,0.08);
    backdrop-filter:blur(8px);
}
.sidebar h3 {
    margin:0 0 12px 0;
    color:var(--accent-dark);
    font-size:15px;
    border-bottom:1px solid #bbf7d0;
    padding-bottom:9px;
}
.sidebar a {
    display:block;
    color:#334155;
    text-decoration:none;
    padding:8px 10px;
    border-radius:10px;
    margin-bottom:5px;
    font-size:13px;
}
.sidebar a:hover {
    background:var(--accent-soft);
    color:var(--accent-dark);
    font-weight:750;
}
.content { min-width:0; }
section {
    background:rgba(255,255,255,0.98);
    border:1px solid var(--border);
    border-radius:18px;
    padding:24px;
    margin-bottom:22px;
    box-shadow:0 10px 28px rgba(15,23,42,0.06);
    overflow-x:auto;
}
section h2 {
    margin:0 0 16px 0;
    color:var(--accent-dark);
    font-size:22px;
    font-weight:850;
    letter-spacing:-0.01em;
    border-bottom:2px solid #bbf7d0;
    padding-bottom:10px;
}
.report-grid {
    display:grid;
    grid-template-columns:repeat(auto-fit, minmax(190px, 1fr));
    gap:14px;
    margin:14px 0;
}
.metric-card {
    background:linear-gradient(180deg, #ffffff, #f8fafc);
    border:1px solid #cbd5e1;
    border-radius:16px;
    padding:16px;
    text-align:center;
    box-shadow:0 6px 18px rgba(15,23,42,0.05);
}
.metric-label { color:var(--muted); font-size:12px; text-transform:uppercase; letter-spacing:0.06em; }
.metric-value { color:var(--accent-dark); font-size:22px; font-weight:850; margin-top:4px; }
.report-table {
    width:100%;
    border-collapse:separate;
    border-spacing:0;
    font-size:13px;
    background:white;
    border:1px solid var(--border);
    border-radius:12px;
    overflow:hidden;
}
.report-table th {
    position:sticky; top:0;
    background:linear-gradient(180deg, #dcfce7, #bbf7d0);
    border-bottom:1px solid #86efac;
    border-right:1px solid #bbf7d0;
    padding:10px;
    text-align:center;
    color:#052e16;
    font-weight:850;
}
.report-table td {
    border-bottom:1px solid #e5e7eb;
    border-right:1px solid #eef2f7;
    padding:9px;
    text-align:center;
    vertical-align:middle;
}
.report-table tr:nth-child(even) td { background:#f8fafc; }
.report-table tr:hover td { background:#f0fdf4; }
.table-wrap { overflow-x:auto; margin:10px 0 14px 0; }
.note {
    background:#f0fdf4;
    border:1px solid #bbf7d0;
    border-left:5px solid var(--accent);
    color:#14532d;
    border-radius:14px;
    padding:14px 16px;
    margin:12px 0 16px 0;
}
.warning {
    background:var(--warning);
    border:1px solid #f59e0b;
    color:#78350f;
    border-radius:14px;
    padding:14px 16px;
    margin:12px 0;
}
.fi-box, .alignment-card {
    background:#f8fafc;
    border:1px solid #cbd5e1;
    border-radius:14px;
    padding:14px;
    margin:12px 0 16px 0;
}
.fi-small { color:#475569; font-size:13px; line-height:1.75; }
.aln-wrap {
    background:#111827;
    border:1px solid #374151;
    border-radius:14px;
    padding:16px 18px;
    overflow-x:auto;
    margin-top:10px;
    box-shadow:inset 0 0 0 1px rgba(255,255,255,0.03);
}
.aln-block { margin-bottom:20px; }
.aln-block-title {
    color:#9ca3af;
    font-size:11px;
    font-family:ui-monospace, SFMono-Regular, Menlo, Monaco, Consolas, "Liberation Mono", monospace;
    margin-bottom:7px;
    letter-spacing:.06em;
    text-transform:uppercase;
}
.aln-row { display:flex; align-items:baseline; line-height:1.58; }
.aln-label {
    color:#d1d5db;
    font-family:ui-monospace, SFMono-Regular, Menlo, Monaco, Consolas, "Liberation Mono", monospace;
    font-size:13px;
    min-width:138px;
    flex-shrink:0;
    overflow:hidden;
    text-overflow:ellipsis;
    white-space:nowrap;
    padding-right:10px;
}
.aln-seq, .aln-match {
    font-family:ui-monospace, SFMono-Regular, Menlo, Monaco, Consolas, "Liberation Mono", monospace;
    font-size:13px;
    white-space:pre;
    letter-spacing:.055em;
}
.aln-match { color:#6b7280; min-width:138px; padding-left:148px; }
.aa-hydrophobic { color:#fb7185; }
.aa-positive    { color:#60a5fa; }
.aa-negative    { color:#fde68a; }
.aa-polar       { color:#86efac; }
.aa-cysteine    { color:#fdba74; }
.aa-glycine     { color:#c4b5fd; }
.aa-proline     { color:#67e8f9; }
.aa-aromatic    { color:#f0abfc; }
.aa-gap         { color:#4b5563; }
.aa-match-id    { color:#86efac; font-weight:850; }
.aa-match-cons  { color:#fde68a; }
.aa-match-none  { color:#4b5563; }
.aln-legend {
    display:flex;
    flex-wrap:wrap;
    gap:10px;
    margin-top:10px;
    font-size:12px;
    font-family:ui-monospace, SFMono-Regular, Menlo, Monaco, Consolas, "Liberation Mono", monospace;
}
.aln-legend-item { display:flex; align-items:center; gap:5px; color:#d1d5db; }
.aln-legend-dot { width:12px; height:12px; border-radius:4px; display:inline-block; }
.footer-note {
    color:#64748b;
    font-size:12px;
    text-align:center;
    padding:8px 0 26px;
}
@media (max-width: 980px) {
    .layout { grid-template-columns:1fr; padding:16px; }
    .sidebar { position:relative; top:auto; }
    header { padding:30px 24px; }
}
@media print {
    .sidebar { display:none; }
    .layout { display:block; padding:0; }
    section { box-shadow:none; break-inside:avoid; }
}
</style>
"""


def df_to_html_report_table(df, max_rows=None):
    """Convert a dataframe to a styled HTML table for standalone reports."""
    if df is None or df.empty:
        return '<p>No data available.</p>'
    out_df = df.copy()
    if max_rows is not None:
        out_df = out_df.head(max_rows)
    return out_df.to_html(
        index=False,
        classes="report-table",
        escape=True,
        border=0
    )

# ============================================================
# Helper functions
# ============================================================

def show_error_message(title, message):
    display(HTML(f"""
    {STYLE_BLOCK}
    <div class="fi-container" style="background-color:#fee2e2;border-color:#f87171;color:#7f1d1d;">
        <h3 style="margin-top:0;">⚠️ {escape(title)}</h3>
        <p style="margin-bottom:0;">{escape(message)}</p>
    </div>{RESIZE_SCRIPT}"""))


def get_mime_type(file_path):
    return {
        ".csv":"text/csv", ".txt":"text/plain", ".log":"text/plain",
        ".fasta":"text/plain", ".fa":"text/plain", ".faa":"text/plain",
        ".json":"application/json", ".html":"text/html", ".htm":"text/html", ".zip":"application/zip",
    }.get(Path(file_path).suffix.lower(), "application/octet-stream")


def make_download_link(file_path, label):
    file_path = Path(file_path)
    if not file_path.exists():
        return ""
    size_mb = file_path.stat().st_size / (1024*1024)
    if size_mb > MAX_DIRECT_DOWNLOAD_MB:
        return (f'<span class="fi-download-note">📁 {escape(label)} ({size_mb:.1f} MB) '
                f'→ <code>{escape(str(file_path))}</code></span>')
    encoded = base64.b64encode(file_path.read_bytes()).decode()
    return (f'<a class="fi-download" download="{escape(file_path.name)}" '
            f'href="data:{get_mime_type(file_path)};base64,{encoded}">⬇️ {escape(label)}</a>')


def create_zip_from_files(file_paths, zip_path):
    zip_path = Path(zip_path)
    with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
        for fp in file_paths:
            fp = Path(fp)
            if fp.exists():
                zf.write(fp, arcname=fp.name)
    return zip_path


def preview_table_html(df, max_rows=7):
    if df is None or df.empty:
        return '<table class="fi-table"><tr><th>Message</th></tr><tr><td>No data.</td></tr></table>'
    return df.head(max_rows).to_html(index=False, classes="fi-table", escape=True)


# ============================================================
# Custom ClustalX-like coloured HTML renderer
# ============================================================

# Fixed residue-class colour groups inspired by ClustalX-like viewers.
# This is intentionally described as ClustalX-like/custom because it does
# not implement the full profile-dependent ClustalX-like colouring algorithm.
_AA_CLASS = {
    **{aa: "aa-hydrophobic" for aa in "AVILM"},
    **{aa: "aa-positive"    for aa in "KRH"},
    **{aa: "aa-negative"    for aa in "DE"},
    **{aa: "aa-polar"       for aa in "STNQ"},
    "C": "aa-cysteine",
    "G": "aa-glycine",
    "P": "aa-proline",
    "-": "aa-gap",
}
# Aromatic override (F W Y)
for aa in "FWY":
    _AA_CLASS[aa] = "aa-aromatic"


def _colour_residue(char):
    cls = _AA_CLASS.get(char.upper(), "")
    if cls:
        return f'<span class="{cls}">{escape(char)}</span>'
    return escape(char)


def _colour_seq(seq):
    return "".join(_colour_residue(c) for c in seq)


def _colour_match(match_line):
    out = []
    for c in match_line:
        if c == "|":
            out.append(f'<span class="aa-match-id">{c}</span>')
        elif c == "+":
            out.append(f'<span class="aa-match-cons">{c}</span>')
        elif c == ".":
            out.append(f'<span class="aa-match-none">{c}</span>')
        else:
            out.append(f'<span class="aa-gap"> </span>')
    return "".join(out)


def build_clustal_html(result, block_width=80):
    """
    Renders a single pairwise alignment result as a coloured custom alignment HTML block.
    Shows: Seq A / match line / Seq B, broken into blocks of block_width.
    """
    name_a    = result["Protein A"]
    name_b    = result["Protein B"]
    aln_a     = result["Reference Alignment"]
    aln_b     = result["Sequence Alignment"]
    match_line= result["Match Pattern"]
    ident     = result["Identity (%)"]
    ident_ng  = result["Identity (% no-gap)"]
    sim       = result["Similarity (%)"]
    gaps      = result["Gaps (%)"]
    score     = result["Score"]
    aln_len   = result["Aligned Length"]

    label_a = escape(name_a[:18].ljust(18))
    label_b = escape(name_b[:18].ljust(18))
    label_m = " " * 18

    blocks_html = []
    for start in range(0, len(aln_a), block_width):
        end   = start + block_width
        seg_a = aln_a[start:end]
        seg_b = aln_b[start:end]
        seg_m = match_line[start:end]

        block = f"""
        <div class="aln-block">
            <div class="aln-block-title">positions {start+1}–{min(end, aln_len)}</div>
            <div class="aln-row">
                <span class="aln-label">{label_a}</span>
                <span class="aln-seq">{_colour_seq(seg_a)}</span>
            </div>
            <div class="aln-row">
                <span class="aln-match">{_colour_match(seg_m)}</span>
            </div>
            <div class="aln-row">
                <span class="aln-label">{label_b}</span>
                <span class="aln-seq">{_colour_seq(seg_b)}</span>
            </div>
        </div>"""
        blocks_html.append(block)

    legend_html = """
    <div class="aln-legend">
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f38ba8"></span>Hydrophobic</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#89b4fa"></span>Positive</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f9e2af"></span>Negative</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#a6e3a1"></span>Polar</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#fab387"></span>Cysteine</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#cba6f7"></span>Glycine</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f5c2e7"></span>Aromatic</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#94e2d5"></span>Proline</div>
        <div class="aln-legend-item"><span style="color:#a6e3a1;font-weight:bold">|</span>&nbsp;Identical</div>
        <div class="aln-legend-item"><span style="color:#f9e2af">+</span>&nbsp;Conservative</div>
        <div class="aln-legend-item"><span style="color:#45475a">.</span>&nbsp;Mismatch</div>
    </div>"""

    stats_bar = (
        f"Score: <b>{score}</b> &nbsp;|&nbsp; "
        f"Identity: <b>{ident}%</b> (no-gap: <b>{ident_ng}%</b>) &nbsp;|&nbsp; "
        f"Similarity: <b>{sim}%</b> &nbsp;|&nbsp; "
        f"Gaps: <b>{gaps}%</b> &nbsp;|&nbsp; "
        f"Length: <b>{aln_len}</b>"
    )

    return f"""
    <div style="margin-bottom:8px;font-size:13px;color:#374151;">{stats_bar}</div>
    <div class="aln-wrap">
        {"".join(blocks_html)}
        {legend_html}
    </div>"""


def build_msa_clustal_html(records, block_width=80):
    """
    Renders a Multiple Sequence Alignment as a coloured custom alignment HTML block.
    Conservation-symbol line: * identical, + conserved group, · variable.
    """
    names = [r.id for r in records]
    seqs  = [str(r.seq) for r in records]
    if not seqs:
        return ""

    aln_len  = len(seqs[0])
    max_nlen = max(len(n) for n in names + ["Conservation"])

    # Build a conservation-symbol line for visualization.
    # Conserved groups (any residue in col belongs to one group → '+')
    _CLUSTAL_GROUPS = [
        set("STA"), set("NEQK"), set("NHQK"), set("NDEQ"),
        set("QHRK"), set("MILV"), set("MILF"), set("HY"), set("FYW"),
    ]

    def conservation_symbol(col):
        non_gap = [c for c in col if c != "-"]
        if not non_gap:
            return " "
        col_set = set(non_gap)
        if len(col_set) == 1 and len(non_gap) == len(col):
            return "*"
        if any(col_set <= g for g in _CLUSTAL_GROUPS):
            return "+"
        return "·"

    conservation_line = "".join(conservation_symbol(col) for col in zip(*seqs))

    def colour_conservation_symbol(c):
        if c == "*":
            return f'<span class="aa-match-id">{c}</span>'
        if c == "+":
            return f'<span class="aa-match-cons">{c}</span>'
        if c == "·":
            return f'<span class="aa-match-none">{c}</span>'
        return f'<span class="aa-gap"> </span>'

    blocks_html = []
    for start in range(0, aln_len, block_width):
        end = min(start + block_width, aln_len)
        rows = ""
        for name, seq in zip(names, seqs):
            label = escape(name[:max_nlen].ljust(max_nlen))
            rows += f"""
            <div class="aln-row">
                <span class="aln-label">{label}</span>
                <span class="aln-seq">{_colour_seq(seq[start:end])}</span>
            </div>"""
        # conservation-symbol row
        cons_seg = "".join(colour_conservation_symbol(c) for c in conservation_line[start:end])
        rows += f"""
        <div class="aln-row">
            <span class="aln-label" style="color:#a6adc8;">{"Conservation".ljust(max_nlen)}</span>
            <span class="aln-seq">{cons_seg}</span>
        </div>"""

        blocks_html.append(f"""
        <div class="aln-block">
            <div class="aln-block-title">positions {start+1}–{end}</div>
            {rows}
        </div>""")

    legend_html = """
    <div class="aln-legend">
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f38ba8"></span>Hydrophobic</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#89b4fa"></span>Positive</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f9e2af"></span>Negative</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#a6e3a1"></span>Polar</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#fab387"></span>Cysteine</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#cba6f7"></span>Glycine</div>
        <div class="aln-legend-item"><span class="aln-legend-dot" style="background:#f5c2e7"></span>Aromatic</div>
        <div class="aln-legend-item"><span style="color:#a6e3a1;font-weight:bold">*</span>&nbsp;Fully conserved</div>
        <div class="aln-legend-item"><span style="color:#f9e2af">+</span>&nbsp;Conserved group</div>
        <div class="aln-legend-item"><span style="color:#45475a">·</span>&nbsp;Variable</div>
    </div>"""

    return f"""
    <div class="aln-wrap">
        {"".join(blocks_html)}
        {legend_html}
    </div>"""


# ============================================================
# Main module
# ============================================================

def run_alignment_module():

    global seq_names, sequences

    # ------------------------------------------------------------
    # Input check
    # ------------------------------------------------------------
    if "sequences" not in globals() or len(sequences) == 0:
        show_error_message(
            "Sequence input is missing",
            "Please run the previous cell first: Protein Sequence / FASTA Input."
        )
        return

    if "seq_names" not in globals() or len(seq_names) != len(sequences):
        seq_names = [f"Seq_{i+1}" for i in range(len(sequences))]

    n_sequences = len(sequences)

    # ------------------------------------------------------------
    # Output directory
    # ------------------------------------------------------------
    if Path("/content/drive/MyDrive").exists():
        PROJECT_DIR = Path("/content/drive/MyDrive/FoldColabFlow_Project")
    else:
        PROJECT_DIR = Path("/content/FoldColabFlow_Project")

    ALIGNMENT_OUTPUT_DIR = PROJECT_DIR / "outputs" / "alignment"
    ALIGNMENT_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    alignment_html_report_path = ALIGNMENT_OUTPUT_DIR / "sequence_alignment_similarity_report.html"

    # ------------------------------------------------------------
    # Utility
    # ------------------------------------------------------------
    VALID_AA = set("ACDEFGHIKLMNPQRSTVWYX")

    def clean_sequence(seq):
        return re.sub(r"\s+", "", str(seq)).upper()

    def validate_sequence(seq):
        return sorted(set(seq) - VALID_AA)

    def safe_name(name, fallback):
        name = re.sub(r"[^A-Za-z0-9_.-]+", "_", str(name).strip()).strip("_")
        return name if name else fallback

    def make_unique_names(names):
        """Ensure stable, unique sequence IDs after sanitisation."""
        counts = Counter()
        unique_names = []
        for name in names:
            counts[name] += 1
            if counts[name] == 1:
                unique_names.append(name)
            else:
                unique_names.append(f"{name}_{counts[name]}")
        return unique_names

    # ------------------------------------------------------------
    # Input normalisation and validation inside this module
    # ------------------------------------------------------------
    cleaned_sequences = []
    invalid_report = []

    for name, seq in zip(seq_names, sequences):
        clean_seq = clean_sequence(seq)
        invalid = validate_sequence(clean_seq)
        if not clean_seq:
            invalid_report.append({
                "Protein": str(name),
                "Issue": "empty sequence"
            })
        elif invalid:
            invalid_report.append({
                "Protein": str(name),
                "Issue": "invalid characters: " + "".join(invalid)
            })
        cleaned_sequences.append(clean_seq)

    if invalid_report:
        raise ValueError(
            "Invalid protein sequences detected in the alignment module: "
            + str(invalid_report)
        )

    sequences = cleaned_sequences
    n_sequences = len(sequences)

    safe_seq_names = make_unique_names(
        [safe_name(n, f"Seq_{i+1}") for i, n in enumerate(seq_names)]
    )

    def write_fasta(names, seqs, output_path):
        with open(output_path, "w", encoding="utf-8") as f:
            for name, seq in zip(names, seqs):
                f.write(f">{name}\n")
                for s in range(0, len(seq), 60):
                    f.write(seq[s:s+60] + "\n")

    # ------------------------------------------------------------
    # Reference selection
    # ------------------------------------------------------------
    reference_warning = ""

    if use_custom_reference_sequence and custom_reference_sequence.strip():
        ref_seq = clean_sequence(custom_reference_sequence)
        invalid = validate_sequence(ref_seq)
        if invalid:
            raise ValueError(f"Custom reference contains invalid characters: {invalid}")
        ref_name = "Custom_Reference"
    else:
        req = safe_name(reference_protein_name, "")
        if req in safe_seq_names:
            idx = safe_seq_names.index(req)
            ref_seq, ref_name = sequences[idx], safe_seq_names[idx]
        else:
            ref_seq, ref_name = sequences[0], safe_seq_names[0]
            reference_warning = (
                f'<p class="fi-small" style="color:#b45309;">⚠️ Reference '
                f'<code>{escape(reference_protein_name)}</code> not found. '
                f'Using <code>{escape(ref_name)}</code>.</p>'
            )

    # ------------------------------------------------------------
    # Aligner setup (Biopython — no manual reconstruction)
    # ------------------------------------------------------------
    blosum62 = substitution_matrices.load("BLOSUM62")

    aligner = PairwiseAligner()
    aligner.mode              = "global"
    aligner.substitution_matrix = blosum62
    aligner.open_gap_score    = -10
    aligner.extend_gap_score  = -0.5

    def run_pairwise_alignment(seq_a, seq_b, name_a, name_b):
        best  = aligner.align(seq_a, seq_b)[0]
        aln_a = str(best[0])
        aln_b = str(best[1])

        aligned_length = len(aln_a)
        identical = conservative_subs = true_mismatches = gap_count = ambiguous_pairs = 0
        match_chars = []

        for a, b in zip(aln_a, aln_b):
            if a == "-" or b == "-":
                gap_count += 1
                match_chars.append(" ")
            elif a == "X" or b == "X":
                # Ambiguous residues are retained and counted separately for transparency.
                ambiguous_pairs += 1
                if COUNT_X_AS_IDENTICAL and a == b:
                    identical += 1
                    match_chars.append("|")
                else:
                    true_mismatches += 1
                    match_chars.append(".")
            elif a == b:
                identical += 1
                match_chars.append("|")
            else:
                try:
                    sc = blosum62[a, b]
                except KeyError:
                    sc = -1
                if sc > 0:
                    conservative_subs += 1
                    match_chars.append("+")
                else:
                    true_mismatches += 1
                    match_chars.append(".")

        match_line    = "".join(match_chars)
        non_gap_length = aligned_length - gap_count

        identity_pct         = (identical / aligned_length)                   * 100 if aligned_length   else 0.0
        identity_pct_nogap   = (identical / non_gap_length)                   * 100 if non_gap_length   else 0.0
        similarity_pct       = ((identical + conservative_subs) / aligned_length)  * 100 if aligned_length   else 0.0
        similarity_pct_nogap = ((identical + conservative_subs) / non_gap_length)  * 100 if non_gap_length   else 0.0
        gap_pct              = (gap_count  / aligned_length)                   * 100 if aligned_length   else 0.0

        return {
            "Comparison":                 f"{name_a} vs {name_b}",
            "Protein A":                  name_a,
            "Protein B":                  name_b,
            "Protein A Length":           len(seq_a),
            "Protein B Length":           len(seq_b),
            "Aligned Length":             aligned_length,
            "Non-Gap Length":             non_gap_length,
            "Score":                      round(best.score, 2),
            "Identity (%)":              round(identity_pct,          2),
            "Identity (% no-gap)":       round(identity_pct_nogap,    2),
            "Similarity (%)":            round(similarity_pct,        2),
            "Similarity (% no-gap)":     round(similarity_pct_nogap,  2),
            "Gaps (%)":                  round(gap_pct,               2),
            "Identical Matches":          identical,
            "Conservative Substitutions": conservative_subs,
            "True Mismatches":            true_mismatches,
            "Ambiguous Residue Pairs":     ambiguous_pairs,
            "Gaps":                       gap_count,
            "Reference Alignment":        aln_a,
            "Match Pattern":              match_line,
            "Sequence Alignment":         aln_b,
        }

    def write_pairwise_alignment_details(results, output_path, block=80):
        with open(output_path, "w", encoding="utf-8") as f:
            for r in results:
                f.write(r["Comparison"] + "\n")
                f.write("=" * 100 + "\n\n")
                aln_a, aln_b, ml = r["Reference Alignment"], r["Sequence Alignment"], r["Match Pattern"]
                for s in range(0, len(ml), block):
                    e = s + block
                    f.write(f"{r['Protein A'][:12].ljust(12)} {aln_a[s:e]}\n")
                    f.write(f"{''.ljust(12)} {ml[s:e]}\n")
                    f.write(f"{r['Protein B'][:12].ljust(12)} {aln_b[s:e]}\n\n")
                f.write("\n\n")

    # ------------------------------------------------------------
    # Header
    # ------------------------------------------------------------
    pairwise_count = n_sequences * (n_sequences - 1) // 2

    display(HTML(f"""
    {STYLE_BLOCK}
    <div class="fi-container">
        <h2 class="fi-title">🔬 Sequence Alignment & Similarity Analysis</h2>
        <div class="fi-box">
            <h3>📍 Current Stage: Sequence Alignment</h3>
            <p style="line-height:1.8;margin-bottom:0;">
                Pairwise alignment: Biopython PairwiseAligner + BLOSUM62 (global, no manual reconstruction).<br>
                MSA: MAFFT + Biopython AlignIO.<br>
                Visualisation: custom ClustalX-like residue-class colour scheme inline in notebook.
            </p>
            {reference_warning}
        </div>
        <div class="fi-box">
            <h3>⚙️ Run Settings</h3>
            <div class="fi-stat-grid">
                <div class="fi-stat-card">
                    <div class="fi-stat-label">Analysis mode</div>
                    <div class="fi-stat-value" style="font-size:15px;">{escape(alignment_analysis_mode)}</div>
                </div>
                <div class="fi-stat-card">
                    <div class="fi-stat-label">Loaded proteins</div>
                    <div class="fi-stat-value">{n_sequences}</div>
                </div>
                <div class="fi-stat-card">
                    <div class="fi-stat-label">Pairwise comparisons</div>
                    <div class="fi-stat-value">{pairwise_count}</div>
                </div>
                <div class="fi-stat-card">
                    <div class="fi-stat-label">Reference protein</div>
                    <div class="fi-stat-value" style="font-size:15px;">{escape(ref_name)}</div>
                </div>
            </div>
        </div>
    </div>{RESIZE_SCRIPT}
    """))

    # ============================================================
    # Initialize outputs
    # ============================================================
    output_files       = []
    all_vs_all_results = []
    reference_results  = []

    all_vs_all_df = reference_df = identity_matrix_df = pd.DataFrame()
    similarity_matrix_df = gap_matrix_df = pd.DataFrame()
    msa_records, msa_summary_df = [], pd.DataFrame()

    pairwise_report_sections_html = ""
    msa_report_sections_html = ""

    all_vs_all_csv = identity_csv = similarity_csv = gap_csv = None
    reference_csv = detailed_alignment_txt = None
    msa_input_fasta = msa_output_fasta = msa_view_txt = msa_summary_csv = None
    software_versions_csv = analysis_settings_csv = None

    # ------------------------------------------------------------
    # Reproducibility metadata
    # ------------------------------------------------------------
    software_versions_csv = ALIGNMENT_OUTPUT_DIR / "software_versions.csv"
    analysis_settings_csv = ALIGNMENT_OUTPUT_DIR / "alignment_analysis_settings.csv"

    def get_command_version(command):
        try:
            completed = subprocess.run(
                command, capture_output=True, text=True, check=False
            )
            version_text = (completed.stdout or completed.stderr or "").strip()
            return version_text.splitlines()[0] if version_text else "not reported"
        except Exception as exc:
            return f"not available ({exc})"

    software_versions_df = pd.DataFrame([
        {"Software": "Python",    "Version": platform.python_version()},
        {"Software": "Biopython", "Version": Bio.__version__},
        {"Software": "pandas",    "Version": pd.__version__},
        {"Software": "MAFFT",     "Version": get_command_version(["mafft", "--version"]) if shutil.which("mafft") else "not installed / not used"},
    ])
    software_versions_df.to_csv(software_versions_csv, index=False)

    analysis_settings_df = pd.DataFrame([
        {"Parameter": "analysis_mode", "Value": alignment_analysis_mode},
        {"Parameter": "pairwise_aligner", "Value": "Bio.Align.PairwiseAligner"},
        {"Parameter": "pairwise_mode", "Value": aligner.mode},
        {"Parameter": "substitution_matrix", "Value": "BLOSUM62"},
        {"Parameter": "gap_model", "Value": "affine"},
        {"Parameter": "open_gap_score", "Value": aligner.open_gap_score},
        {"Parameter": "extend_gap_score", "Value": aligner.extend_gap_score},
        {"Parameter": "terminal_gap_handling", "Value": "Biopython PairwiseAligner default handling under the configured global affine gap scores"},
        {"Parameter": "pairwise_alignment_selection", "Value": "first highest-scoring alignment returned by aligner.align(...)[0]"},
        {"Parameter": "identity_full_denominator", "Value": "all alignment columns"},
        {"Parameter": "identity_no_gap_denominator", "Value": "columns with residues in both sequences"},
        {"Parameter": "similarity_definition", "Value": "identical residues + non-identical non-gap pairs with positive BLOSUM62 score"},
        {"Parameter": "gap_count_definition", "Value": "alignment columns containing a gap in either sequence"},
        {"Parameter": "ambiguous_X_pairwise_policy", "Value": "X retained; exact X-X positions are counted as identical by string identity" if COUNT_X_AS_IDENTICAL else "X retained but not counted as identical"},
        {"Parameter": "msa_tool", "Value": "MAFFT"},
        {"Parameter": "mafft_strategy", "Value": "--auto --amino"},
        {"Parameter": "visualisation", "Value": "custom ClustalX-like fixed residue-class colour scheme"},
    ])
    analysis_settings_df.to_csv(analysis_settings_csv, index=False)
    output_files.extend([software_versions_csv, analysis_settings_csv])

    # ============================================================
    # All-vs-All + Reference Summary
    # ============================================================
    run_pairwise_block = alignment_analysis_mode in [
        "All-vs-All Pairwise + Reference Summary", "Both"
    ]

    if run_pairwise_block:
        if n_sequences < 2:
            display(HTML(f"""
            {STYLE_BLOCK}
            <div class="fi-container">
                <div class="fi-box">
                    <h3 style="color:#b45309;">⚠️ Pairwise analysis skipped</h3>
                    <p style="margin-bottom:0;">At least 2 proteins are required.</p>
                </div>
            </div>{RESIZE_SCRIPT}"""))
            pairwise_report_sections_html = """
            <section id="pairwise-summary">
                <h2>Pairwise &amp; Reference Summary</h2>
                <div class="warning">Pairwise analysis was skipped because at least 2 proteins are required.</div>
            </section>"""
        else:
            identity_matrix_df   = pd.DataFrame(100.0, index=safe_seq_names, columns=safe_seq_names)
            similarity_matrix_df = pd.DataFrame(100.0, index=safe_seq_names, columns=safe_seq_names)
            gap_matrix_df        = pd.DataFrame(0.0,   index=safe_seq_names, columns=safe_seq_names)

            for i, j in itertools.combinations(range(n_sequences), 2):
                result = run_pairwise_alignment(
                    sequences[i], sequences[j],
                    safe_seq_names[i], safe_seq_names[j]
                )
                all_vs_all_results.append(result)
                for mat, key in [
                    (identity_matrix_df,   "Identity (%)"),
                    (similarity_matrix_df, "Similarity (%)"),
                    (gap_matrix_df,        "Gaps (%)"),
                ]:
                    mat.loc[safe_seq_names[i], safe_seq_names[j]] = result[key]
                    mat.loc[safe_seq_names[j], safe_seq_names[i]] = result[key]

            all_vs_all_df = pd.DataFrame(all_vs_all_results)

            summary_cols = [
                "Comparison", "Score",
                "Identity (%)", "Identity (% no-gap)",
                "Similarity (%)", "Similarity (% no-gap)",
                "Gaps (%)",
                "Identical Matches", "Conservative Substitutions",
                "True Mismatches", "Ambiguous Residue Pairs", "Gaps"
            ]
            all_vs_all_summary_df = all_vs_all_df[summary_cols]

            all_vs_all_csv        = ALIGNMENT_OUTPUT_DIR / "all_vs_all_pairwise_summary.csv"
            identity_csv          = ALIGNMENT_OUTPUT_DIR / "identity_matrix.csv"
            similarity_csv        = ALIGNMENT_OUTPUT_DIR / "similarity_matrix.csv"
            gap_csv               = ALIGNMENT_OUTPUT_DIR / "gap_matrix.csv"
            detailed_alignment_txt= ALIGNMENT_OUTPUT_DIR / "all_vs_all_pairwise_alignment_details.txt"

            all_vs_all_df.to_csv(all_vs_all_csv, index=False)
            identity_matrix_df.to_csv(identity_csv)
            similarity_matrix_df.to_csv(similarity_csv)
            gap_matrix_df.to_csv(gap_csv)
            write_pairwise_alignment_details(all_vs_all_results, detailed_alignment_txt)
            output_files.extend([all_vs_all_csv, identity_csv, similarity_csv, gap_csv, detailed_alignment_txt])

            mean_identity   = all_vs_all_df["Identity (%)"].mean()
            mean_similarity = all_vs_all_df["Similarity (%)"].mean()
            mean_gaps       = all_vs_all_df["Gaps (%)"].mean()
            closest_row     = all_vs_all_df.loc[all_vs_all_df["Identity (%)"].idxmax()]
            distant_row     = all_vs_all_df.loc[all_vs_all_df["Identity (%)"].idxmin()]

            top_pairs_html = preview_table_html(
                all_vs_all_summary_df.sort_values("Identity (%)", ascending=False),
                TOP_PAIRWISE_ROWS_TO_SHOW
            )

            preview_names       = safe_seq_names[:IDENTITY_MATRIX_PREVIEW_LIMIT]
            identity_preview_html = (
                identity_matrix_df.loc[preview_names, preview_names]
                .round(2).to_html(classes="fi-table", escape=True)
            )

            # Reference summary
            for i, seq in enumerate(sequences):
                r = run_pairwise_alignment(ref_seq, seq, ref_name, safe_seq_names[i])
                r["Reference Protein"] = ref_name
                r["Target Protein"]    = safe_seq_names[i]
                reference_results.append(r)

            reference_df  = pd.DataFrame(reference_results)
            reference_csv = ALIGNMENT_OUTPUT_DIR / "reference_based_summary.csv"
            reference_df.to_csv(reference_csv, index=False)
            output_files.append(reference_csv)

            ref_without_self = reference_df[reference_df["Target Protein"] != ref_name]
            if len(ref_without_self) > 0:
                most_sim  = ref_without_self.loc[ref_without_self["Identity (%)"].idxmax()]
                least_sim = ref_without_self.loc[ref_without_self["Identity (%)"].idxmin()]
                most_ref_text  = f"{most_sim['Target Protein']} ({most_sim['Identity (%)']}%)"
                least_ref_text = f"{least_sim['Target Protein']} ({least_sim['Identity (%)']}%)"
            else:
                most_ref_text = least_ref_text = "Not available"

            # ── Build CLUSTAL viewer: reference vs all others ──────────────
            ref_aln_blocks = ""
            for r in reference_results:
                if r["Target Protein"] == ref_name:
                    continue
                ref_aln_blocks += f"""
                <div class="fi-box" style="margin-bottom:14px;">
                    <h4 style="margin-bottom:6px;">
                        🧬 {escape(r["Protein A"])} vs {escape(r["Protein B"])}
                    </h4>
                    {build_clustal_html(r, block_width=MSA_BLOCK_WIDTH)}
                </div>"""

            reference_report_cols = [
                "Target Protein", "Score",
                "Identity (%)", "Identity (% no-gap)",
                "Similarity (%)", "Similarity (% no-gap)",
                "Gaps (%)", "Identical Matches",
                "Conservative Substitutions", "True Mismatches", "Ambiguous Residue Pairs", "Gaps"
            ]
            reference_report_cols = [c for c in reference_report_cols if c in reference_df.columns]

            identity_report_df = (
                identity_matrix_df.round(2)
                .reset_index()
                .rename(columns={"index": "Protein"})
            )
            similarity_report_df = (
                similarity_matrix_df.round(2)
                .reset_index()
                .rename(columns={"index": "Protein"})
            )
            gap_report_df = (
                gap_matrix_df.round(2)
                .reset_index()
                .rename(columns={"index": "Protein"})
            )
            ref_aln_blocks_report = ref_aln_blocks.replace('class="fi-box"', 'class="alignment-card"')

            pairwise_report_sections_html = f"""
            <section id="pairwise-summary">
                <h2>Pairwise &amp; Reference Summary</h2>
                <div class="report-grid">
                    <div class="metric-card">
                        <div class="metric-label">Mean identity</div>
                        <div class="metric-value">{mean_identity:.2f}%</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Mean similarity</div>
                        <div class="metric-value">{mean_similarity:.2f}%</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Mean gaps</div>
                        <div class="metric-value">{mean_gaps:.2f}%</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Comparisons</div>
                        <div class="metric-value">{len(all_vs_all_df)}</div>
                    </div>
                </div>
                <div class="note">
                    <b>Closest pair:</b> {escape(str(closest_row["Comparison"]))} — {closest_row["Identity (%)"]}%<br>
                    <b>Most distant pair:</b> {escape(str(distant_row["Comparison"]))} — {distant_row["Identity (%)"]}%<br>
                    <b>Most similar to reference:</b> {escape(most_ref_text)}<br>
                    <b>Least similar to reference:</b> {escape(least_ref_text)}
                </div>
                <h3>All-vs-All Summary</h3>
                <div class="table-wrap">
                    {df_to_html_report_table(all_vs_all_summary_df.sort_values("Identity (%)", ascending=False))}
                </div>
            </section>

            <section id="reference-summary">
                <h2>Reference-Based Summary</h2>
                <div class="note">
                    Reference protein: <b>{escape(ref_name)}</b>. Full-alignment identity and gap-excluded identity are reported separately.
                </div>
                <div class="table-wrap">
                    {df_to_html_report_table(reference_df[reference_report_cols])}
                </div>
            </section>

            <section id="alignment-matrices">
                <h2>Identity / Similarity / Gap Matrices</h2>
                <div class="note">
                    Matrix values are percentages rounded to two decimals. Full CSV files are also saved in the output folder.
                </div>
                <h3>Identity Matrix (%)</h3>
                <div class="table-wrap">{df_to_html_report_table(identity_report_df)}</div>
                <h3>Similarity Matrix (%)</h3>
                <div class="table-wrap">{df_to_html_report_table(similarity_report_df)}</div>
                <h3>Gap Matrix (%)</h3>
                <div class="table-wrap">{df_to_html_report_table(gap_report_df)}</div>
            </section>

            <section id="reference-alignments">
                <h2>Reference Alignments — ClustalX-like Colour View</h2>
                <div class="note">
                    Each target sequence is aligned against <b>{escape(ref_name)}</b>. Residues are coloured by fixed residue-class groups.
                </div>
                {ref_aln_blocks_report if ref_aln_blocks_report.strip() else "<p>No reference alignment blocks available.</p>"}
            </section>
            """

            display(HTML(f"""
            {STYLE_BLOCK}
            <div class="fi-container">
                <h3 style="color:#374151;margin-top:0;">🔹 Pairwise & Reference Summary</h3>

                <div class="fi-box">
                    <h4>📊 Compact Results</h4>
                    <div class="fi-stat-grid">
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Mean identity (aligned)</div>
                            <div class="fi-stat-value">{mean_identity:.2f}%</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Mean similarity (aligned)</div>
                            <div class="fi-stat-value">{mean_similarity:.2f}%</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Mean gaps</div>
                            <div class="fi-stat-value">{mean_gaps:.2f}%</div>
                        </div>
                    </div>
                    <p style="line-height:1.8;margin-top:14px;margin-bottom:0;">
                        <b>Closest pair:</b> {escape(str(closest_row["Comparison"]))} — {closest_row["Identity (%)"]}%<br>
                        <b>Most distant pair:</b> {escape(str(distant_row["Comparison"]))} — {distant_row["Identity (%)"]}%<br>
                        <b>Most similar to reference:</b> {escape(most_ref_text)}<br>
                        <b>Least similar to reference:</b> {escape(least_ref_text)}
                    </p>
                </div>

                <div class="fi-box">
                    <h4>📊 All-vs-All Summary (top {TOP_PAIRWISE_ROWS_TO_SHOW})</h4>
                    <p class="fi-small">
                        <b>Identity (%)</b> = over full aligned length (gaps included) &nbsp;|&nbsp;
                        <b>Identity (% no-gap)</b> = gap-excluded identity over non-gap aligned columns
                    </p>
                    {top_pairs_html}
                    {make_download_link(all_vs_all_csv, "Download full All-vs-All CSV")}
                </div>

                <div class="fi-box">
                    <h4>🧬 Identity Matrix Preview (%)</h4>
                    <p class="fi-small">First {min(IDENTITY_MATRIX_PREVIEW_LIMIT, n_sequences)} proteins shown.</p>
                    {identity_preview_html}
                    {make_download_link(identity_csv,   "Identity Matrix CSV")}
                    {make_download_link(similarity_csv, "Similarity Matrix CSV")}
                    {make_download_link(gap_csv,        "Gap Matrix CSV")}
                </div>

                <div class="fi-box">
                    <h4>🎨 Reference Alignments — ClustalX-like Colour View</h4>
                    <p class="fi-small">
                        Each protein aligned against <b>{escape(ref_name)}</b>.
                        Residues are coloured by fixed residue-class groups (ClustalX-like scheme).
                    </p>
                    {ref_aln_blocks}
                </div>

            </div>{RESIZE_SCRIPT}
            """))

    # ============================================================
    # MAFFT MSA — Biopython AlignIO + AlignInfo + coloured view
    # ============================================================
    run_msa_block = alignment_analysis_mode in [
        "Multiple Sequence Alignment with MAFFT", "Both"
    ]

    if run_msa_block:
        if n_sequences < 2:
            display(HTML(f"""
            {STYLE_BLOCK}
            <div class="fi-container">
                <div class="fi-box">
                    <h3 style="color:#b45309;">⚠️ MSA skipped</h3>
                    <p style="margin-bottom:0;">At least 2 proteins are required.</p>
                </div>
            </div>{RESIZE_SCRIPT}"""))
            msa_report_sections_html = """
            <section id="msa-summary">
                <h2>Multiple Sequence Alignment</h2>
                <div class="warning">MSA was skipped because at least 2 proteins are required.</div>
            </section>"""
        else:
            if shutil.which("mafft") is None:
                raise RuntimeError("MAFFT is not installed or not found in PATH.")

            msa_input_fasta  = ALIGNMENT_OUTPUT_DIR / "mafft_input_sequences.fasta"
            msa_output_fasta = ALIGNMENT_OUTPUT_DIR / "mafft_alignment.fasta"
            msa_view_txt     = ALIGNMENT_OUTPUT_DIR / "mafft_alignment_view.txt"
            msa_summary_csv  = ALIGNMENT_OUTPUT_DIR / "mafft_alignment_summary.csv"

            write_fasta(safe_seq_names, sequences, msa_input_fasta)

            with open(msa_output_fasta, "w", encoding="utf-8") as out_f:
                subprocess.run(
                    ["mafft", "--auto", "--amino", str(msa_input_fasta)],
                    stdout=out_f, stderr=subprocess.PIPE, text=True, check=True
                )

            # Biopython AlignIO reads the MSA
            msa_alignment = AlignIO.read(str(msa_output_fasta), "fasta")
            msa_records   = list(msa_alignment)
            msa_alignment_length = msa_alignment.get_alignment_length()
            # No consensus sequence is generated. The MSA is summarized using
            # per-sequence gap metrics and column-level conservation statistics.

            # Per-sequence stats
            msa_summary_rows = []
            for record in msa_records:
                aln_seq     = str(record.seq)
                gap_count   = aln_seq.count("-")
                non_gap_len = len(aln_seq) - gap_count
                gap_pct     = (gap_count / len(aln_seq)) * 100 if aln_seq else 0.0
                msa_summary_rows.append({
                    "Protein Name":    record.id,
                    "Original Length": non_gap_len,
                    "MSA Length":      len(aln_seq),
                    "Gap Count":       gap_count,
                    "Gap (%)":         round(gap_pct, 2),
                })

            # Column conservation stats
            conserved_columns = gap_columns = variable_columns = 0
            n_seqs = len(msa_records)
            for col_idx in range(msa_alignment_length):
                col      = [str(rec.seq)[col_idx] for rec in msa_records]
                non_gaps = [c for c in col if c != "-"]
                if "-" in set(col):
                    gap_columns += 1
                if len(non_gaps) == n_seqs and len(set(non_gaps)) == 1:
                    conserved_columns += 1
                else:
                    variable_columns  += 1

            conserved_pct  = (conserved_columns / msa_alignment_length) * 100 if msa_alignment_length else 0.0
            gap_column_pct = (gap_columns       / msa_alignment_length) * 100 if msa_alignment_length else 0.0

            msa_summary_df = pd.DataFrame(msa_summary_rows)
            msa_summary_df.to_csv(msa_summary_csv, index=False)

            # Plain-text MSA view (TXT file)
            def build_msa_text(records, block_width=80):
                names    = [r.id for r in records]
                seqs     = [str(r.seq) for r in records]
                aln_len  = len(seqs[0]) if seqs else 0
                max_nlen = max(len(n) for n in names + ["Conservation"])
                lines    = []
                for start in range(0, aln_len, block_width):
                    end = min(start + block_width, aln_len)
                    lines.append(f"Block {start+1}-{end}")
                    lines.append("-" * (max_nlen + block_width + 2))
                    for name, seq in zip(names, seqs):
                        lines.append(f"{name.ljust(max_nlen)}  {seq[start:end]}")
                    lines.append("")
                return "\n".join(lines)

            with open(msa_view_txt, "w", encoding="utf-8") as f:
                f.write(build_msa_text(msa_records, MSA_BLOCK_WIDTH))

            output_files.extend([
                msa_input_fasta, msa_output_fasta, msa_view_txt,
                msa_summary_csv
            ])

            # ── Coloured MSA HTML ──────────────────────────────────────────
            msa_coloured_html = build_msa_clustal_html(msa_records, block_width=MSA_BLOCK_WIDTH)

            msa_report_sections_html = f"""
            <section id="msa-summary">
                <h2>Multiple Sequence Alignment (MAFFT)</h2>
                <div class="report-grid">
                    <div class="metric-card">
                        <div class="metric-label">Tool</div>
                        <div class="metric-value" style="font-size:20px;">MAFFT</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Proteins</div>
                        <div class="metric-value">{len(msa_records)}</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">MSA length</div>
                        <div class="metric-value">{msa_alignment_length}</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Conserved columns</div>
                        <div class="metric-value">{conserved_pct:.2f}%</div>
                    </div>
                    <div class="metric-card">
                        <div class="metric-label">Gap columns</div>
                        <div class="metric-value">{gap_column_pct:.2f}%</div>
                    </div>
                </div>
                <div class="note">
                    Column-level statistics summarize fully conserved, gap-containing and variable alignment columns.
                </div>
                <h3>Per-Sequence Gap Summary</h3>
                <div class="table-wrap">{df_to_html_report_table(msa_summary_df)}</div>
            </section>

            <section id="msa-clustal-view">
                <h2>MSA — ClustalX-like Colour View</h2>
                <div class="note">
                    Conservation-symbol line: <b>*</b> fully conserved, <b>+</b> conserved group, <b>·</b> variable, blank when gaps dominate.
                </div>
                {msa_coloured_html}
            </section>
            """

            display(HTML(f"""
            {STYLE_BLOCK}
            <div class="fi-container">
                <h3 style="color:#374151;margin-top:0;">🔹 Multiple Sequence Alignment (MAFFT)</h3>

                <div class="fi-box">
                    <div class="fi-stat-grid">
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Tool</div>
                            <div class="fi-stat-value">MAFFT</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Proteins</div>
                            <div class="fi-stat-value">{len(msa_records)}</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">MSA length</div>
                            <div class="fi-stat-value">{msa_alignment_length}</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Conserved columns</div>
                            <div class="fi-stat-value">{conserved_pct:.2f}%</div>
                        </div>
                        <div class="fi-stat-card">
                            <div class="fi-stat-label">Gap columns</div>
                            <div class="fi-stat-value">{gap_column_pct:.2f}%</div>
                        </div>
                    </div>
                </div>

                <div class="fi-box">
                    <h4>📋 Per-Sequence Gap Summary</h4>
                    {df_to_html_report_table(msa_summary_df)}
                </div>

                <div class="fi-box">
                    <h4>🎨 MSA — ClustalX-like Colour View</h4>
                    <p class="fi-small">
                        Residues coloured by physicochemical group (ClustalX-like scheme).<br>
                        Conservation-symbol line: <b style="color:#a6e3a1">*</b> fully conserved &nbsp;
                        <b style="color:#f9e2af">+</b> conserved group &nbsp;
                        <b style="color:#45475a">·</b> variable &nbsp;
                        <b style="color:#45475a">&nbsp;</b> gap present
                    </p>
                    {msa_coloured_html}
                </div>

            </div>{RESIZE_SCRIPT}
            """))

    # ============================================================
    # Standalone HTML Report
    # ============================================================
    if export_alignment_html_report:
        report_nav_links = ""
        available_sections = pairwise_report_sections_html + msa_report_sections_html
        for section_id, section_label in [
            ("pairwise-summary", "Pairwise Summary"),
            ("reference-summary", "Reference Summary"),
            ("alignment-matrices", "Matrices"),
            ("reference-alignments", "Reference Alignments"),
            ("msa-summary", "MSA Summary"),
            ("msa-clustal-view", "MSA Clustal View"),
        ]:
            if f'id="{section_id}"' in available_sections:
                report_nav_links += f'<a href="#{section_id}">{section_label}</a>\n'
        if not report_nav_links.strip():
            report_nav_links = '<a href="#overview">Overview</a>'

        generated_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        alignment_report_html = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>FoldColabFlow Sequence Alignment & Similarity Analysis Report</title>
{ALIGNMENT_REPORT_CSS}
</head>
<body>
<header>
    <h1>🔬 FoldColabFlow Sequence Alignment &amp; Similarity Analysis Report</h1>
    <p>{n_sequences} protein(s) loaded | Analysis mode: {escape(alignment_analysis_mode)} | Reference: {escape(ref_name)}</p>
</header>

<div class="layout">
<aside class="sidebar">
    <h3>Navigation</h3>
    <a href="#overview">Overview</a>
    {report_nav_links}
    <a href="#outputs">Output Files</a>
</aside>

<main class="content">
<section id="overview">
    <h2>Overview</h2>
    <div class="report-grid">
        <div class="metric-card">
            <div class="metric-label">Loaded proteins</div>
            <div class="metric-value">{n_sequences}</div>
        </div>
        <div class="metric-card">
            <div class="metric-label">Pairwise comparisons</div>
            <div class="metric-value">{pairwise_count}</div>
        </div>
        <div class="metric-card">
            <div class="metric-label">Reference</div>
            <div class="metric-value" style="font-size:18px;">{escape(ref_name)}</div>
        </div>
        <div class="metric-card">
            <div class="metric-label">Block width</div>
            <div class="metric-value">{MSA_BLOCK_WIDTH}</div>
        </div>
    </div>
    <div class="note">
        Pairwise alignment uses Biopython PairwiseAligner with BLOSUM62 in global mode.
        Multiple sequence alignment uses MAFFT when enabled.
        The standalone report preserves the custom ClustalX-like coloured alignment viewers.
    </div>
</section>

{pairwise_report_sections_html if pairwise_report_sections_html.strip() else ""}
{msa_report_sections_html if msa_report_sections_html.strip() else ""}

<section id="outputs">
    <h2>Output Files</h2>
    <div class="note">
        Report generated: <b>{generated_at}</b><br>
        Output directory: <code>{escape(str(ALIGNMENT_OUTPUT_DIR))}</code>
    </div>
    <div class="table-wrap">
        {df_to_html_report_table(pd.DataFrame([
            {"File": Path(fp).name, "Type": Path(fp).suffix.lower() or "file", "Size (KB)": round(Path(fp).stat().st_size / 1024, 2)}
            for fp in output_files if Path(fp).exists()
        ]))}
    </div>
</section>
</main>
</div>
<div class="footer-note">Generated by FoldColabFlow</div>
</body>
</html>"""

        with open(alignment_html_report_path, "w", encoding="utf-8") as f:
            f.write(alignment_report_html)
        output_files.append(alignment_html_report_path)

    # ============================================================
    # ZIP + Download links
    # ============================================================
    zip_path = None
    if output_files:
        zip_path = ALIGNMENT_OUTPUT_DIR / "alignment_outputs.zip"
        create_zip_from_files(output_files, zip_path)

    links = ""
    for path, label in [
        (zip_path,                "Download all outputs as ZIP"),
        (alignment_html_report_path if export_alignment_html_report else None,
                                  "Sequence Alignment HTML Report"),
        (all_vs_all_csv,          "All-vs-All CSV"),
        (identity_csv,            "Identity Matrix CSV"),
        (similarity_csv,          "Similarity Matrix CSV"),
        (gap_csv,                 "Gap Matrix CSV"),
        (reference_csv,           "Reference Summary CSV"),
        (detailed_alignment_txt,  "Pairwise Details TXT"),
        (msa_output_fasta,        "MAFFT Alignment FASTA"),
        (msa_summary_csv,         "MSA Summary CSV"),
        (msa_view_txt,            "MSA Alignment TXT"),
    ]:
        if path is not None:
            links += make_download_link(path, label)

    display(HTML(f"""
    {STYLE_BLOCK}
    <div class="fi-container">
        <h3 style="color:#374151;margin-top:0;">✅ Alignment Module Completed</h3>
        <div class="fi-box">
            <h4>📁 Output Files</h4>
            <p style="line-height:1.8;margin-bottom:10px;">
                Saved in: <code>{escape(str(ALIGNMENT_OUTPUT_DIR))}</code>
            </p>
            {links}
        </div>
        <div class="fi-next">
            <p style="margin:0;font-size:16px;">
                ➡️ <b>Next stage:</b> Physico-Chemical Properties Analysis
            </p>
        </div>
    </div>{RESIZE_SCRIPT}
    """))

    # ============================================================
    # Save globals for downstream cells
    # ============================================================
    globals().update({
        "all_vs_all_results":   all_vs_all_results,
        "all_vs_all_df":        all_vs_all_df,
        "reference_results":    reference_results,
        "reference_df":         reference_df,
        "identity_matrix_df":   identity_matrix_df,
        "similarity_matrix_df": similarity_matrix_df,
        "gap_matrix_df":        gap_matrix_df,
        "msa_records":          msa_records,
        "msa_summary_df":       msa_summary_df,
        "alignment_output_dir": str(ALIGNMENT_OUTPUT_DIR),
        "alignment_report_file": (
            str(alignment_html_report_path)
            if export_alignment_html_report and alignment_html_report_path.exists()
            else None
        ),
        "software_versions_file": str(software_versions_csv) if software_versions_csv else None,
        "alignment_analysis_settings_file": str(analysis_settings_csv) if analysis_settings_csv else None,
        # backward-compat
        "alignment_results":    all_vs_all_results,
        "alignment_df":         all_vs_all_df,
    })

    if msa_output_fasta:
        globals()["mafft_alignment_file"] = str(msa_output_fasta)
    if zip_path:
        globals()["alignment_zip_file"] = str(zip_path)

    if "save_state" in globals():
        save_state(
            "02_sequence_alignment",
            all_vs_all_results=all_vs_all_results,
            all_vs_all_df=all_vs_all_df,
            reference_results=reference_results,
            reference_df=reference_df,
            identity_matrix_df=identity_matrix_df,
            similarity_matrix_df=similarity_matrix_df,
            gap_matrix_df=gap_matrix_df,
            msa_summary_df=msa_summary_df,
            alignment_output_dir=str(ALIGNMENT_OUTPUT_DIR),
            alignment_report_file=(
                str(alignment_html_report_path)
                if export_alignment_html_report and alignment_html_report_path.exists()
                else None
            ),
            mafft_alignment_file=str(msa_output_fasta) if msa_output_fasta else None,
            software_versions_file=str(software_versions_csv) if software_versions_csv else None,
            alignment_analysis_settings_file=str(analysis_settings_csv) if analysis_settings_csv else None,
            alignment_zip_file=str(zip_path) if zip_path else None,
        )


# ============================================================
# Run
# ============================================================
run_alignment_module()
```

```python
#@title 🧪 Physico-Chemical Properties Analysis { display-mode: "form" }

show_table = True #@param {type:"boolean"}
show_box_violin = True #@param {type:"boolean"}
show_heatmap = True #@param {type:"boolean"}
show_parallel = True #@param {type:"boolean"}
show_scatter_matrix = True #@param {type:"boolean"}
show_stats_ranking = True #@param {type:"boolean"}
export_html_report = True #@param {type:"boolean"}

# ============================================================
# Physico-Chemical Properties Analysis
# FoldColabFlow optimized version
# FIXES in this version:
#   1. fig.show() / display(fig) added for all plots so they
#      actually render inside Colab output (was missing before)
#   2. Parallel coordinates color_column bug fixed:
#      Solubility Proxy Score (%) is now kept un-normalized
#      in a separate column so color encoding works correctly
#   3. All other logic unchanged
#   4. Publication typography styling added: black plot text/labels only; data colors unchanged
#   5. High-resolution PNG/SVG export added for manuscript-quality figure downloads
#   6. Correlation heatmap now excludes constant/zero-variance features to avoid blank rows/columns
# ============================================================

# ====================== INSTALLATIONS ======================

import sys
import subprocess

base_packages = ["biopython", "pandas"]
plot_requested = show_box_violin or show_heatmap or show_parallel or show_scatter_matrix

if plot_requested:
    install_packages = base_packages + ["matplotlib", "plotly", "kaleido==0.2.1"]
else:
    install_packages = base_packages

subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q"] + install_packages,
    check=True
)

# ====================== IMPORTS ======================

from Bio.SeqUtils.ProtParam import ProteinAnalysis
import pandas as pd
from pathlib import Path
from html import escape
from IPython.display import HTML, display
import base64
import zipfile
import json
import re
import math

if plot_requested:
    import matplotlib.pyplot as plt
    import plotly.express as px
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots

# ============================================================
# Internal Settings
# ============================================================

STANDARD_AA = set("ACDEFGHIKLMNPQRSTVWY")
AA_ORDER = list("ACDEFGHIKLMNPQRSTVWY")

TABLE_PREVIEW_ROWS = 10
MAX_DIRECT_DOWNLOAD_MB = 30

MIN_ANALYZABLE_LENGTH = 20
VERY_LONG_SEQUENCE_WARNING = 1500

INSTABILITY_STABLE_THRESHOLD = 40
GRAVY_HYDROPHILIC_THRESHOLD = -0.4
GRAVY_HYDROPHOBIC_THRESHOLD = 0.4
SOLUBILITY_ABS_CHARGE_THRESHOLD = 4.5
HYDROPHOBIC_RESIDUES = set("IVLFMAC")

SOLUBILITY_POINTS_STABILITY = 40
SOLUBILITY_POINTS_HYDROPHILICITY = 40
SOLUBILITY_POINTS_CHARGE = 20
SOLUBILITY_MAX_POINTS = (
    SOLUBILITY_POINTS_STABILITY
    + SOLUBILITY_POINTS_HYDROPHILICITY
    + SOLUBILITY_POINTS_CHARGE
)

LOW_SOLUBILITY_PROXY_THRESHOLD = 40
HIGH_SOLUBILITY_PROXY_THRESHOLD = 70

MIN_N_FOR_DISTRIBUTION_PLOT = 2
MIN_N_FOR_CORRELATION = 6
MIN_N_FOR_PARALLEL = 3
MIN_N_FOR_SCATTER = 6

# ============================================================
# Project Directory
# ============================================================

if Path("/content/drive/MyDrive").exists():
    PROJECT_DIR = Path("/content/drive/MyDrive/FoldColabFlow_Project")
else:
    PROJECT_DIR = Path("/content/FoldColabFlow_Project")

PHYCHEM_OUTPUT_DIR = PROJECT_DIR / "outputs" / "physchem"
PHYCHEM_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ============================================================
# Colab Output Styling
# ============================================================

STYLE_BLOCK = """
<style>
.fi-container {
    background-color:#e5e7eb;
    border:2px solid #9ca3af;
    border-radius:16px;
    padding:22px;
    font-family:Arial, sans-serif;
    color:#1f2937;
    margin:16px 0;
    overflow:visible;
    max-height:none;
}
.fi-title {
    text-align:center;
    color:#1d4ed8;
    margin-top:0;
    margin-bottom:18px;
    font-weight:900;
}
.fi-box {
    background-color:#f8fafc;
    border:1px solid #cbd5e1;
    border-radius:12px;
    padding:14px;
    margin-bottom:16px;
}
.fi-box h3, .fi-box h4 {
    color:#111827;
    margin-top:0;
    font-weight:900;
}
.fi-warning-box {
    background-color:#fee2e2;
    border:2px solid #ef4444;
    border-radius:12px;
    padding:14px;
    margin-bottom:16px;
}
.fi-warning-box h3, .fi-warning-box h4 {
    color:#991b1b;
    margin-top:0;
    font-weight:900;
}
.fi-next {
    background-color:#dcfce7;
    border:1px solid #86efac;
    border-radius:12px;
    padding:14px;
    margin-top:16px;
    color:#166534;
}
.fi-table {
    width:100%;
    border-collapse:collapse;
    font-size:13px;
    margin:10px 0 12px 0;
    background-color:#ffffff;
    table-layout:auto;
}
.fi-table th {
    background-color:#dbeafe;
    color:#111827;
    text-align:center !important;
    padding:9px;
    border:1px solid #94a3b8;
    font-weight:900;
    white-space:normal;
}
.fi-table td {
    background-color:#ffffff;
    color:#1f2937;
    text-align:center !important;
    padding:8px;
    border:1px solid #cbd5e1;
    vertical-align:middle;
    white-space:normal;
}
.fi-table tr:nth-child(even) td { background-color:#f8fafc; }
.fi-small {
    font-size:13px;
    color:#374151;
    line-height:1.7;
    font-weight:600;
}
.fi-download {
    display:inline-block;
    background-color:#eff6ff;
    border:1px solid #93c5fd;
    color:#1d4ed8;
    padding:8px 12px;
    border-radius:10px;
    text-decoration:none;
    font-weight:bold;
    margin:4px 6px 4px 0;
    font-size:13px;
}
.fi-download:hover { background-color:#dbeafe; }
.fi-download-note {
    display:inline-block;
    background-color:#f1f5f9;
    border:1px solid #cbd5e1;
    color:#475569;
    padding:8px 12px;
    border-radius:10px;
    margin:4px 6px 4px 0;
    font-size:13px;
}
.fi-stat-grid {
    display:grid;
    grid-template-columns:repeat(auto-fit, minmax(170px, 1fr));
    gap:12px;
    margin-top:10px;
}
.fi-stat-card {
    background-color:#ffffff;
    border:1px solid #cbd5e1;
    border-radius:12px;
    padding:12px;
    text-align:center;
}
.fi-stat-label { font-size:13px; color:#111827; margin-bottom:6px; font-weight:900; }
.fi-stat-value { font-size:18px; font-weight:900; color:#1f2937; }
.fi-method {
    background-color:#eef2ff;
    border:1px solid #c7d2fe;
    border-radius:12px;
    padding:12px;
    color:#312e81;
    line-height:1.8;
    font-weight:600;
}
.fi-min-cell { background-color:#fee2e2 !important; border:2px solid #f87171 !important; font-weight:900; }
.fi-max-cell { background-color:#dcfce7 !important; border:2px solid #86efac !important; font-weight:900; }
.fi-same-cell { background-color:#fef9c3 !important; border:2px solid #fde68a !important; font-weight:900; }
</style>
"""

RESIZE_SCRIPT = """
<script>
  try {
    google.colab.output.setIframeHeight(0, true, {maxHeight: 1000000});
  } catch(e) {}
</script>
"""

# ============================================================
# HTML Report Styling
# ============================================================

REPORT_CSS = """
<style>
:root {
    --blue:#1d4ed8;
    --blue-dark:#1e3a8a;
    --blue-soft:#dbeafe;
    --blue-line:#93c5fd;
    --slate:#111827;
    --muted:#64748b;
    --border:#d1d5db;
    --bg:#f3f4f6;
    --card:#ffffff;
    --note:#eef2ff;
    --warning:#fee2e2;
}
* { box-sizing:border-box; }
html { scroll-behavior:smooth; }
body {
    font-family:Inter, ui-sans-serif, system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", Arial, Helvetica, sans-serif;
    background:
        radial-gradient(circle at top left, rgba(37,99,235,0.11), transparent 34%),
        linear-gradient(180deg, #f8fafc 0%, var(--bg) 100%);
    color:var(--slate);
    margin:0;
    padding:0;
    line-height:1.65;
}
header {
    background:linear-gradient(135deg, #0f172a 0%, #1e3a8a 55%, #1d4ed8 100%);
    color:white;
    padding:38px 46px;
    border-bottom:5px solid #1e40af;
    box-shadow:0 14px 34px rgba(15,23,42,0.16);
}
header h1 { margin:0 0 8px 0; font-size:31px; line-height:1.2; font-weight:850; letter-spacing:-0.02em; }
header p  { margin:0; opacity:0.92; font-size:14px; }
.layout {
    display:grid;
    grid-template-columns:280px minmax(0, 1fr);
    gap:24px;
    padding:26px;
    max-width:1720px;
    margin:0 auto;
}
.sidebar {
    position:sticky; top:18px; align-self:start;
    background:rgba(255,255,255,0.94);
    border:1px solid var(--border);
    border-radius:18px;
    padding:18px;
    box-shadow:0 10px 26px rgba(15,23,42,0.08);
    backdrop-filter:blur(8px);
    max-height:calc(100vh - 36px);
    overflow-y:auto;
    overscroll-behavior:contain;
    scrollbar-gutter:stable;
}
.sidebar h3 {
    margin:0 0 12px 0;
    font-size:15px;
    color:var(--blue-dark);
    border-bottom:1px solid var(--blue-soft);
    padding-bottom:9px;
}
.sidebar a {
    display:block;
    text-decoration:none;
    color:#334155;
    padding:8px 10px;
    border-radius:10px;
    font-size:13px;
    margin-bottom:5px;
}
.sidebar a:hover { background:#eff6ff; color:var(--blue-dark); font-weight:750; }
.content { min-width:0; }
section {
    background:rgba(255,255,255,0.98);
    border-radius:18px;
    padding:24px;
    margin-bottom:22px;
    border:1px solid var(--border);
    box-shadow:0 10px 28px rgba(15,23,42,0.06);
    overflow-x:auto;
}
section h2 {
    margin:0 0 16px 0;
    color:var(--blue-dark);
    font-size:22px;
    font-weight:850;
    letter-spacing:-0.01em;
    border-bottom:2px solid var(--blue-soft);
    padding-bottom:10px;
}
.report-table {
    width:100%;
    border-collapse:separate;
    border-spacing:0;
    font-size:13px;
    background:white;
    border:1px solid var(--border);
    border-radius:12px;
    overflow:hidden;
}
.report-table th {
    position:sticky; top:0;
    background:linear-gradient(180deg, #dbeafe, #bfdbfe);
    border-bottom:1px solid #93c5fd;
    border-right:1px solid #bfdbfe;
    padding:10px;
    text-align:center;
    color:#111827;
    font-weight:850;
}
.report-table td {
    border-bottom:1px solid #e5e7eb;
    border-right:1px solid #eef2f7;
    padding:9px;
    text-align:center;
    vertical-align:middle;
}
.report-table tr:nth-child(even) td { background:#f8fafc; }
.report-table tr:hover td { background:#eff6ff; }
.note {
    background:var(--note);
    border:1px solid #c7d2fe;
    border-left:5px solid var(--blue);
    color:#312e81;
    border-radius:14px;
    padding:14px 16px;
    line-height:1.8;
}
.warning {
    background:var(--warning);
    border:1px solid #ef4444;
    color:#7f1d1d;
    border-radius:14px;
    padding:14px 16px;
    line-height:1.8;
}
.academic-note {
    background:#f8fafc;
    border-left:5px solid #2563eb;
    padding:12px 14px;
    margin-bottom:16px;
    color:#334155;
    border-radius:10px;
}
img { max-width:100%; height:auto; border:1px solid #d1d5db; border-radius:14px; }
.plotly-graph-div {
    border:1px solid #e5e7eb;
    border-radius:14px;
    overflow:hidden;
    box-shadow:0 6px 18px rgba(15,23,42,0.04);
}
@media (max-width: 980px) {
    .layout { grid-template-columns:1fr; padding:16px; }
    .sidebar { position:relative; top:auto; max-height:none; overflow-y:visible; }
    header { padding:30px 24px; }
}
@media print {
    .sidebar { display:none; }
    .layout { display:block; padding:0; }
    section { box-shadow:none; break-inside:avoid; }
}
</style>
"""

# ============================================================
# Helper Functions
# ============================================================

def show_error_message(title, message):
    display(HTML(f"""
    {STYLE_BLOCK}
    <div class="fi-container">
        <div class="fi-warning-box">
            <h3>❌ {escape(title)}</h3>
            <p style="margin-bottom:0;">{escape(message)}</p>
        </div>
    </div>
    {RESIZE_SCRIPT}
    """))


def get_mime_type(file_path):
    suffix = Path(file_path).suffix.lower()
    return {
        ".csv":  "text/csv",
        ".txt":  "text/plain",
        ".log":  "text/plain",
        ".json": "application/json",
        ".html": "text/html",
        ".png":  "image/png",
        ".zip":  "application/zip",
    }.get(suffix, "application/octet-stream")


def make_download_link(file_path, label):
    file_path = Path(file_path)
    if not file_path.exists():
        return ""
    file_size_mb = file_path.stat().st_size / (1024 * 1024)
    if file_size_mb > MAX_DIRECT_DOWNLOAD_MB:
        return (
            f'<span class="fi-download-note">📁 {escape(label)} '
            f'({file_size_mb:.1f} MB) saved at '
            f'<code>{escape(str(file_path))}</code></span>'
        )
    encoded = base64.b64encode(file_path.read_bytes()).decode("utf-8")
    return (
        f'<a class="fi-download" download="{escape(file_path.name)}" '
        f'href="data:{get_mime_type(file_path)};base64,{encoded}">'
        f'⬇️ {escape(label)}</a>'
    )


def create_zip_from_files(file_paths, zip_path):
    zip_path = Path(zip_path)
    with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:
        for fp in file_paths:
            fp = Path(fp)
            if fp.exists():
                zipf.write(fp, arcname=fp.name)
    return zip_path


def safe_name(name, fallback):
    name = re.sub(r"[^A-Za-z0-9_.-]+", "_", str(name).strip()).strip("_")
    return name if name else fallback


def clean_for_protparam(seq):
    original = str(seq).upper()
    cleaned  = "".join(aa for aa in original if aa in STANDARD_AA)
    removed  = sorted(set(original) - STANDARD_AA)
    return cleaned, removed


def amino_acid_counts(seq):
    return {aa: seq.count(aa) for aa in AA_ORDER}


def amino_acid_percentages(seq):
    length = len(seq)
    if length == 0:
        return {aa: 0 for aa in AA_ORDER}
    return {aa: round((seq.count(aa) / length) * 100, 3) for aa in AA_ORDER}


def calculate_aliphatic_index(seq):
    """
    Aliphatic Index as defined by Ikai (1980).
    AI = Ala% + 2.9 × Val% + 3.9 × (Ile% + Leu%)
    Ref: Ikai AJ (1980) J Biochem 88(6):1895-1898.
    """
    length = len(seq)
    if length == 0:
        return 0.0
    ala = (seq.count("A") / length) * 100
    val = (seq.count("V") / length) * 100
    ile = (seq.count("I") / length) * 100
    leu = (seq.count("L") / length) * 100
    return round(ala + 2.9 * val + 3.9 * (ile + leu), 2)


def classify_instability(instability):
    return "Stable" if instability < INSTABILITY_STABLE_THRESHOLD else "Potentially unstable"


def classify_gravy(gravy):
    if gravy < GRAVY_HYDROPHILIC_THRESHOLD:
        return "Hydrophilic"
    if gravy > GRAVY_HYDROPHOBIC_THRESHOLD:
        return "Hydrophobic"
    return "Neutral / mixed"


def classify_pi(pi):
    if pi < 6:
        return "Acidic"
    if pi > 8:
        return "Basic"
    return "Near-neutral"


def classify_solubility_proxy(score):
    if score >= HIGH_SOLUBILITY_PROXY_THRESHOLD:
        return "High proxy"
    if score >= LOW_SOLUBILITY_PROXY_THRESHOLD:
        return "Moderate proxy"
    return "Low proxy"


def calculate_solubility_proxy_score(instability, gravy, charge):
    raw_score = 0
    criteria  = []
    if instability < INSTABILITY_STABLE_THRESHOLD:
        raw_score += SOLUBILITY_POINTS_STABILITY
        criteria.append("Instability Index < 40: passed (+40 pts)")
    else:
        criteria.append("Instability Index < 40: not passed")
    if gravy < GRAVY_HYDROPHILIC_THRESHOLD:
        raw_score += SOLUBILITY_POINTS_HYDROPHILICITY
        criteria.append("GRAVY < -0.4: passed (+40 pts)")
    else:
        criteria.append("GRAVY < -0.4: not passed")
    if abs(charge) >= SOLUBILITY_ABS_CHARGE_THRESHOLD:
        raw_score += SOLUBILITY_POINTS_CHARGE
        criteria.append(
            f"|Net charge at pH7| >= {SOLUBILITY_ABS_CHARGE_THRESHOLD}: passed (+20 pts)"
        )
    else:
        criteria.append(
            f"|Net charge at pH7| >= {SOLUBILITY_ABS_CHARGE_THRESHOLD}: not passed"
        )
    score = round((raw_score / SOLUBILITY_MAX_POINTS) * 100, 1)
    return score, "; ".join(criteria)


def add_warning(warning_list, protein_name, warning_type, details, recommendation):
    warning_list.append({
        "Protein Name":   protein_name,
        "Warning Type":   warning_type,
        "Details":        details,
        "Recommendation": recommendation
    })


def add_plot_note(plot_notes, plot_name, status, message):
    plot_notes.append({"Plot": plot_name, "Status": status, "Message": message})


def add_plot_caution(plot_cautions, plot_name, message):
    plot_cautions.append({"Plot / Analysis": plot_name, "Caution": message})


def format_cell_value(value):
    if value is None:
        return ""
    if isinstance(value, float):
        if math.isnan(value):
            return ""
        return f"{value:.3f}".rstrip("0").rstrip(".")
    return str(value)


def highlight_min_max_table_html(df, max_rows=10):
    if df is None or df.empty:
        return (
            '<table class="fi-table"><tr><th>Message</th></tr>'
            '<tr><td>No data available.</td></tr></table>'
        )
    preview_df   = df.head(max_rows).copy()
    numeric_cols = preview_df.select_dtypes(include="number").columns.tolist()
    min_values = {
        col: preview_df[col].dropna().min()
        for col in numeric_cols if len(preview_df[col].dropna()) > 0
    }
    max_values = {
        col: preview_df[col].dropna().max()
        for col in numeric_cols if len(preview_df[col].dropna()) > 0
    }
    html = '<table class="fi-table"><thead><tr>'
    for col in preview_df.columns:
        html += f"<th>{escape(str(col))}</th>"
    html += "</tr></thead><tbody>"
    for _, row in preview_df.iterrows():
        html += "<tr>"
        for col in preview_df.columns:
            value      = row[col]
            cell_class = ""
            if col in numeric_cols and col in min_values and col in max_values:
                if pd.notna(value):
                    if min_values[col] == max_values[col]:
                        cell_class = "fi-same-cell"
                    elif value == max_values[col]:
                        cell_class = "fi-max-cell"
                    elif value == min_values[col]:
                        cell_class = "fi-min-cell"
            html += f'<td class="{cell_class}">{escape(format_cell_value(value))}</td>'
        html += "</tr>"
    html += "</tbody></table>"
    if len(df) > max_rows:
        html += (
            f'<p class="fi-small">Showing first {max_rows} rows. '
            f'Download the HTML/ZIP report for complete results.</p>'
        )
    html += (
        '<p class="fi-small">Color guide: '
        '🟩 maximum, 🟥 minimum, 🟨 same min/max value in the displayed preview.</p>'
    )
    return html


def preview_table_html(df, max_rows=10):
    if df is None or df.empty:
        return (
            '<table class="fi-table"><tr><th>Message</th></tr>'
            '<tr><td>No data available.</td></tr></table>'
        )
    html = df.head(max_rows).to_html(index=False, classes="fi-table", escape=True)
    if len(df) > max_rows:
        html += (
            f'<p class="fi-small">Showing first {max_rows} rows. '
            f'Download the report for complete results.</p>'
        )
    return html


def df_to_html_report_table(df):
    if df is None or df.empty:
        return "<p>No data available.</p>"
    return df.to_html(
        index=False, border=0, classes="report-table",
        justify="center", escape=True
    )


# ============================================================
# Publication Plot Typography / Label Styling
# ============================================================
# This block changes ONLY plot typography, tick/legend/colorbar text,
# and spacing. It does NOT modify axis lines, zero-lines, violin/box/scatter/heatmap
# fill colors, marker colors, data values, thresholds, calculations, or normalization.

PLOT_FONT_FAMILY = "Arial, sans-serif"
PLOT_TEXT_COLOR = "black"
PLOT_AXIS_COLOR = "black"
PLOT_BACKGROUND_COLOR = "white"


def apply_publication_text_axis_style(
    fig,
    *,
    width=1100,
    height=650,
    title_size=22,
    x_title=None,
    y_title=None,
    tickangle=None,
    showlegend=None,
    left_margin=90,
    right_margin=120,
    top_margin=120,
    bottom_margin=110,
    colorbar_title=None,
):
    """
    Manuscript-friendly Plotly styling.

    Only these visual elements are changed:
    - title, axis titles, tick labels, legend text, colorbar text -> black
    - margins/spacing -> expanded to avoid overlap

    Axis lines, zero-lines, data-encoding colors, marker colors,
    and plot fills are intentionally left unchanged or disabled where they
    only add extra visual lines.
    """
    fig.update_layout(
        template="plotly_white",
        width=width,
        height=height,
        paper_bgcolor=PLOT_BACKGROUND_COLOR,
        font=dict(
            family=PLOT_FONT_FAMILY,
            color=PLOT_TEXT_COLOR,
            size=14,
        ),
        title=dict(
            x=0.5,
            xanchor="center",
            y=0.965,
            yanchor="top",
            font=dict(size=title_size, color=PLOT_TEXT_COLOR),
            pad=dict(t=8, b=22),
        ),
        margin=dict(
            l=left_margin,
            r=right_margin,
            t=top_margin,
            b=bottom_margin,
        ),
        legend=dict(
            orientation="v",
            x=1.03,
            xanchor="left",
            y=1.0,
            yanchor="top",
            bgcolor="rgba(255,255,255,0)",
            font=dict(size=13, color=PLOT_TEXT_COLOR),
            title_font=dict(size=13, color=PLOT_TEXT_COLOR),
            tracegroupgap=8,
        ),
    )

    if showlegend is not None:
        fig.update_layout(showlegend=showlegend)

    # Axis/tick/title TEXT only. Do not add black axis baselines or zero-lines.
    # This prevents the extra black horizontal/vertical lines seen in some outputs.
    axis_style = dict(
        showline=False,
        zeroline=False,
        ticks="",
        tickfont=dict(color=PLOT_TEXT_COLOR, size=12),
        title_font=dict(color=PLOT_TEXT_COLOR, size=14),
        automargin=True,
    )
    fig.update_xaxes(**axis_style)
    fig.update_yaxes(**axis_style)

    if x_title is not None:
        fig.update_xaxes(title_text=x_title)
    if y_title is not None:
        fig.update_yaxes(title_text=y_title)
    if tickangle is not None:
        fig.update_xaxes(tickangle=tickangle)

    # Colorbar text only. The color scale itself is not changed.
    colorbar_text_style = dict(
        tickfont=dict(color=PLOT_TEXT_COLOR, size=12),
        title=dict(
            text=colorbar_title if colorbar_title else None,
            font=dict(color=PLOT_TEXT_COLOR, size=13),
            side="right",
        ),
        x=1.005,
        xpad=4,
        len=0.82,
    )
    fig.update_coloraxes(colorbar=colorbar_text_style)

    # Trace-specific text containers; no marker/fill/color-scale changes.
    for trace in fig.data:
        trace_type = getattr(trace, "type", None)

        if trace_type == "heatmap":
            trace.update(textfont=dict(color=PLOT_TEXT_COLOR, size=12))

        if trace_type == "parcoords":
            trace.update(
                labelfont=dict(color=PLOT_TEXT_COLOR, size=12),
                tickfont=dict(color=PLOT_TEXT_COLOR, size=11),
                rangefont=dict(color=PLOT_TEXT_COLOR, size=11),
            )
            if hasattr(trace, "line") and hasattr(trace.line, "colorbar"):
                trace.line.colorbar.update(colorbar_text_style)

        if hasattr(trace, "marker") and hasattr(trace.marker, "colorbar"):
            try:
                trace.marker.colorbar.update(colorbar_text_style)
            except Exception:
                pass

    # Annotation text, used in some Plotly compound figures.
    try:
        fig.for_each_annotation(
            lambda annotation: annotation.update(
                font=dict(color=PLOT_TEXT_COLOR, size=12)
            )
        )
    except Exception:
        pass

    return fig


# ============================================================
# High-Resolution Figure Export Helper
# ============================================================
# Static figure files are saved in addition to interactive HTML files.
# SVG is vector-based and preferred for manuscripts when the journal accepts it.
# PNG is exported at high scale for raster workflows.

STATIC_EXPORT_SCALE = 4


def save_plotly_figure_outputs(
    fig,
    *,
    base_name,
    output_dir,
    output_files,
    plot_notes,
    plot_label,
):
    """
    Save one Plotly figure as:
    - interactive HTML
    - high-resolution PNG
    - SVG vector file

    This function changes no data and no visual encodings. It only exports files.
    """
    output_dir = Path(output_dir)

    html_path = output_dir / f"{base_name}.html"
    fig.write_html(str(html_path), include_plotlyjs="cdn")
    output_files.append(html_path)
    add_plot_note(plot_notes, plot_label, "Generated HTML", str(html_path))

    export_width = int(fig.layout.width) if fig.layout.width else 1200
    export_height = int(fig.layout.height) if fig.layout.height else 800

    png_path = output_dir / f"{base_name}_highres.png"
    svg_path = output_dir / f"{base_name}.svg"

    try:
        fig.write_image(
            str(png_path),
            format="png",
            width=export_width,
            height=export_height,
            scale=STATIC_EXPORT_SCALE,
        )
        fig.write_image(
            str(svg_path),
            format="svg",
            width=export_width,
            height=export_height,
        )
        output_files.extend([png_path, svg_path])
        add_plot_note(
            plot_notes,
            f"{plot_label} static export",
            "Generated PNG/SVG",
            f"{png_path}; {svg_path}",
        )
    except Exception as e:
        add_plot_note(
            plot_notes,
            f"{plot_label} static export",
            "Failed",
            (
                "Static image export failed. Interactive HTML was still saved. "
                "In Colab, rerun after installing kaleido. Error: " + str(e)
            ),
        )

    return html_path


# ============================================================
# Main Module
# ============================================================

if "sequences" not in globals() or len(sequences) == 0:
    show_error_message(
        "No sequences found",
        "Please run the Protein Sequence / FASTA Input cell before this analysis."
    )

else:

    if "seq_names" not in globals() or len(seq_names) != len(sequences):
        seq_names = [f"Seq_{i+1}" for i in range(len(sequences))]

    results         = []
    skipped_records  = []
    warning_records  = []
    aa_count_rows    = []
    aa_percent_rows  = []
    plot_notes       = []
    plot_cautions    = []

    # ============================================================
    # Analysis Loop
    # ============================================================

    for i, raw_seq in enumerate(sequences):
        original_name   = seq_names[i]
        protein_name    = safe_name(original_name, f"Seq_{i+1}")
        original_seq    = str(raw_seq).upper()
        original_length = len(original_seq)

        cleaned_seq, removed_chars = clean_for_protparam(original_seq)
        analyzed_length = len(cleaned_seq)

        if removed_chars:
            add_warning(
                warning_records, protein_name,
                "Non-standard residues excluded",
                f"Excluded residues for ProtParam calculation: {', '.join(removed_chars)}.",
                "Calculations are based only on the 20 standard amino acids."
            )

        if analyzed_length < MIN_ANALYZABLE_LENGTH:
            skipped_records.append({
                "Protein Name":    protein_name,
                "Original Length": original_length,
                "Analyzed Length": analyzed_length,
                "Reason": (
                    f"Analyzed sequence length is below {MIN_ANALYZABLE_LENGTH} amino acids."
                )
            })
            continue

        if analyzed_length > VERY_LONG_SEQUENCE_WARNING:
            add_warning(
                warning_records, protein_name,
                "Very long protein",
                f"Analyzed length is {analyzed_length} amino acids.",
                "Very long proteins may require more time and memory in downstream structure prediction."
            )

        try:
            analysis    = ProteinAnalysis(cleaned_seq)
            length      = analyzed_length
            mw          = analysis.molecular_weight()
            pi          = analysis.isoelectric_point()
            gravy       = analysis.gravy()
            charge      = analysis.charge_at_pH(7.0)
            instability = analysis.instability_index()
            aromaticity = analysis.aromaticity()
            try:
                extinction_reduced, extinction_oxidized = analysis.molar_extinction_coefficient()
            except Exception:
                extinction_reduced, extinction_oxidized = None, None
        except Exception as e:
            skipped_records.append({
                "Protein Name":    protein_name,
                "Original Length": original_length,
                "Analyzed Length": analyzed_length,
                "Reason": f"ProtParam calculation failed: {str(e)}"
            })
            continue

        aa_counts   = amino_acid_counts(cleaned_seq)
        aa_percents = amino_acid_percentages(cleaned_seq)
        aa_count_row = {"Protein Name": protein_name}
        aa_count_row.update(aa_counts)
        aa_count_rows.append(aa_count_row)
        aa_percent_row = {"Protein Name": protein_name}
        aa_percent_row.update(aa_percents)
        aa_percent_rows.append(aa_percent_row)

        hydrophobic_count = sum(cleaned_seq.count(aa) for aa in HYDROPHOBIC_RESIDUES)
        hydrophobic_pct   = round((hydrophobic_count / length) * 100, 2)
        hydrophilic_pct   = round(100 - hydrophobic_pct, 2)
        charged_count     = sum(cleaned_seq.count(aa) for aa in "KRDE")
        charged_pct       = round((charged_count / length) * 100, 2)
        acidic_count      = cleaned_seq.count("D") + cleaned_seq.count("E")
        acidic_pct        = round((acidic_count / length) * 100, 2)
        basic_count       = cleaned_seq.count("K") + cleaned_seq.count("R")
        basic_pct         = round((basic_count / length) * 100, 2)
        cysteine_count    = cleaned_seq.count("C")
        cysteine_pct      = round((cysteine_count / length) * 100, 2)
        charge_density    = round(charge / length, 4)
        aliphatic_index   = calculate_aliphatic_index(cleaned_seq)

        instability_class = classify_instability(instability)
        gravy_class       = classify_gravy(gravy)
        pi_class          = classify_pi(pi)
        solubility_proxy_score, solubility_proxy_criteria = calculate_solubility_proxy_score(
            instability=instability, gravy=gravy, charge=charge
        )
        solubility_proxy_class = classify_solubility_proxy(solubility_proxy_score)

        if instability >= INSTABILITY_STABLE_THRESHOLD:
            add_warning(
                warning_records, protein_name, "Potentially unstable",
                f"Instability Index = {instability:.2f}.",
                "Instability Index above 40 is a theoretical sequence-based instability flag."
            )
        if gravy > GRAVY_HYDROPHOBIC_THRESHOLD:
            add_warning(
                warning_records, protein_name, "Hydrophobic tendency",
                f"GRAVY = {gravy:.3f}.",
                "Positive/high GRAVY indicates hydrophobic tendency."
            )
        if solubility_proxy_score < LOW_SOLUBILITY_PROXY_THRESHOLD:
            add_warning(
                warning_records, protein_name, "Low solubility proxy",
                f"Solubility Proxy Score = {solubility_proxy_score}%.",
                "Heuristic screening flag only. Use Protein-Sol, SoluProt, or CamSol for validation."
            )
        if pi < 5 or pi > 9:
            add_warning(
                warning_records, protein_name, "Extreme theoretical pI",
                f"pI = {pi:.2f}.",
                "Very acidic or basic pI may affect purification strategy."
            )
        if cysteine_count >= 4:
            add_warning(
                warning_records, protein_name, "High cysteine count",
                f"Cysteine count = {cysteine_count}.",
                "Multiple cysteines may indicate possible disulfide bonding."
            )

        results.append({
            "Protein Name":                protein_name,
            "Original Length":             original_length,
            "Analyzed Length":             length,
            "Mol. Weight (Da)":            round(mw, 2),
            "Mol. Weight (kDa)":           round(mw / 1000, 3),
            "pI":                          round(pi, 2),
            "pI Class":                    pi_class,
            "GRAVY":                       round(gravy, 3),
            "GRAVY Class":                 gravy_class,
            "Net Charge (pH7)":            round(charge, 2),
            "Charge Density":              charge_density,
            "Instability Index":           round(instability, 2),
            "Instability Class":           instability_class,
            "Aromaticity":                 round(aromaticity, 3),
            "Aliphatic Index":             aliphatic_index,
            "Extinction Coeff. Reduced":   extinction_reduced,
            "Extinction Coeff. Oxidized":  extinction_oxidized,
            "Hydrophobic (%)":             hydrophobic_pct,
            "Hydrophilic (%)":             hydrophilic_pct,
            "Charged Residues (%)":        charged_pct,
            "Acidic Residues (%)":         acidic_pct,
            "Basic Residues (%)":          basic_pct,
            "Cysteine Count":              cysteine_count,
            "Cysteine (%)":                cysteine_pct,
            "Solubility Proxy Score (%)":  solubility_proxy_score,
            "Solubility Proxy Class":      solubility_proxy_class,
            "Solubility Proxy Criteria":   solubility_proxy_criteria
        })

    # ============================================================
    # DataFrames
    # ============================================================

    physchem_df         = pd.DataFrame(results)
    physchem_warning_df = pd.DataFrame(warning_records)
    physchem_skipped_df = pd.DataFrame(skipped_records)
    aa_count_df         = pd.DataFrame(aa_count_rows)
    aa_percent_df       = pd.DataFrame(aa_percent_rows)

    if physchem_df.empty:
        msg = "No sequences could be analyzed."
        if not physchem_skipped_df.empty:
            msg += " Check the skipped records table."
        show_error_message("Physico-chemical analysis failed", msg)

    else:

        core_columns = [
            "Protein Name", "Analyzed Length", "Mol. Weight (kDa)",
            "pI", "GRAVY", "Net Charge (pH7)", "Instability Index",
            "Instability Class", "Solubility Proxy Score (%)", "Solubility Proxy Class"
        ]

        additional_columns = [
            "Protein Name", "Aliphatic Index",
            "Extinction Coeff. Reduced", "Extinction Coeff. Oxidized",
            "Charged Residues (%)", "Cysteine Count", "Cysteine (%)",
            "Charge Density", "GRAVY Class", "pI Class"
        ]

        numeric_summary_columns = [
            "Analyzed Length", "Mol. Weight (kDa)", "pI", "GRAVY",
            "Net Charge (pH7)", "Charge Density", "Instability Index",
            "Aromaticity", "Aliphatic Index", "Hydrophobic (%)",
            "Hydrophilic (%)", "Charged Residues (%)", "Cysteine Count",
            "Solubility Proxy Score (%)"
        ]

        core_preview_df       = physchem_df[core_columns].copy()
        additional_preview_df = physchem_df[additional_columns].copy()

        # ============================================================
        # Output Files
        # ============================================================

        output_files = []

        physchem_csv     = PHYCHEM_OUTPUT_DIR / "physicochemical_properties_full.csv"
        core_csv         = PHYCHEM_OUTPUT_DIR / "physicochemical_core_columns.csv"
        additional_csv   = PHYCHEM_OUTPUT_DIR / "physicochemical_additional_indicators.csv"
        warnings_csv     = PHYCHEM_OUTPUT_DIR / "physicochemical_warnings.csv"
        skipped_csv      = PHYCHEM_OUTPUT_DIR / "physicochemical_skipped_sequences.csv"
        aa_count_csv     = PHYCHEM_OUTPUT_DIR / "amino_acid_composition_counts.csv"
        aa_percent_csv   = PHYCHEM_OUTPUT_DIR / "amino_acid_composition_percentages.csv"
        methodology_json = PHYCHEM_OUTPUT_DIR / "physicochemical_methodology.json"
        plot_notes_csv   = PHYCHEM_OUTPUT_DIR / "physicochemical_plot_notes.csv"
        plot_caution_csv = PHYCHEM_OUTPUT_DIR / "physicochemical_plot_cautions.csv"
        html_report_path = PHYCHEM_OUTPUT_DIR / "physicochemical_report.html"

        physchem_df.to_csv(physchem_csv, index=False)
        core_preview_df.to_csv(core_csv, index=False)
        additional_preview_df.to_csv(additional_csv, index=False)
        physchem_warning_df.to_csv(warnings_csv, index=False)
        physchem_skipped_df.to_csv(skipped_csv, index=False)
        aa_count_df.to_csv(aa_count_csv, index=False)
        aa_percent_df.to_csv(aa_percent_csv, index=False)
        output_files.extend([
            physchem_csv, core_csv, additional_csv,
            warnings_csv, skipped_csv, aa_count_csv, aa_percent_csv
        ])

        methodology = {
            "Core package": "Biopython Bio.SeqUtils.ProtParam.ProteinAnalysis",
            "Instability Index": {
                "Method": "Guruprasad et al. (1990) dipeptide instability weights",
                "Threshold": "< 40 = Stable",
                "References": [
                    "Guruprasad K et al. (1990) Protein Eng 4(2):155-161",
                    "DOI: 10.2174/0929866526666190228144219"
                ]
            },
            "GRAVY": {
                "Method": "Kyte-Doolittle hydrophobicity scale, mean over residues",
                "Thresholds": {"Hydrophilic": "GRAVY < -0.4", "Hydrophobic": "GRAVY > 0.4"},
                "References": [
                    "Kyte J & Doolittle RF (1982) J Mol Biol 157(1):105-132",
                    "DOI: 10.3390/ijms24032481"
                ]
            },
            "Aliphatic Index": {
                "Formula": "AI = Ala% + 2.9*Val% + 3.9*(Ile%+Leu%)",
                "References": ["Ikai AJ (1980) J Biochem 88(6):1895-1898"]
            },
            "Hydrophobic residue set": {
                "Residues": "I, V, L, F, M, A, C  (KD score > 0)",
                "References": ["Kyte J & Doolittle RF (1982) J Mol Biol 157(1):105-132"]
            },
            "Solubility Proxy Score": {
                "Type": "Heuristic screening score — NOT a validated solubility prediction model",
                "Criteria and weights": {
                    "Instability Index < 40": f"+{SOLUBILITY_POINTS_STABILITY} pts",
                    "GRAVY < -0.4":           f"+{SOLUBILITY_POINTS_HYDROPHILICITY} pts",
                    f"|Net charge at pH7| >= {SOLUBILITY_ABS_CHARGE_THRESHOLD}":
                        f"+{SOLUBILITY_POINTS_CHARGE} pts"
                },
                "Total possible points": SOLUBILITY_MAX_POINTS,
                "Weight rationale": "Weights are internal design choices without independent validation."
            },
            "Charge Density": {
                "Formula": "Net charge at pH7 / sequence length",
                "Note": "Derived metric for size-normalized charge comparison."
            },
            "Plot skip policy": (
                "All plots are generated with >= 2 analyzed proteins. "
                "Caution notes are added for small N but plots are never skipped."
            )
        }

        with open(methodology_json, "w", encoding="utf-8") as f:
            json.dump(methodology, f, indent=2, ensure_ascii=False)
        output_files.append(methodology_json)

        # ============================================================
        # Summary Statistics
        # ============================================================

        analyzed_count = len(physchem_df)
        skipped_count  = len(physchem_skipped_df)
        warning_count  = len(physchem_warning_df)

        stable_count         = int((physchem_df["Instability Class"] == "Stable").sum())
        unstable_count       = int((physchem_df["Instability Class"] == "Potentially unstable").sum())
        hydrophilic_count_s  = int((physchem_df["GRAVY Class"] == "Hydrophilic").sum())
        hydrophobic_count_s  = int((physchem_df["GRAVY Class"] == "Hydrophobic").sum())
        high_sol_proxy_count = int((physchem_df["Solubility Proxy Class"] == "High proxy").sum())
        low_sol_proxy_count  = int((physchem_df["Solubility Proxy Class"] == "Low proxy").sum())

        most_hydrophobic         = physchem_df.loc[physchem_df["GRAVY"].idxmax(), "Protein Name"]
        most_hydrophilic         = physchem_df.loc[physchem_df["GRAVY"].idxmin(), "Protein Name"]
        most_stable              = physchem_df.loc[physchem_df["Instability Index"].idxmin(), "Protein Name"]
        highest_proxy_solubility = physchem_df.loc[
            physchem_df["Solubility Proxy Score (%)"].idxmax(), "Protein Name"
        ]

        physchem_summary = {
            "Analyzed proteins":            analyzed_count,
            "Skipped proteins":             skipped_count,
            "Warning count":                warning_count,
            "Average length":               round(physchem_df["Analyzed Length"].mean(), 2),
            "Average molecular weight kDa": round(physchem_df["Mol. Weight (kDa)"].mean(), 3),
            "Average pI":                   round(physchem_df["pI"].mean(), 2),
            "Average GRAVY":                round(physchem_df["GRAVY"].mean(), 3),
            "Average solubility proxy":     round(physchem_df["Solubility Proxy Score (%)"].mean(), 1),
            "Stable count":                 stable_count,
            "Potentially unstable count":   unstable_count,
            "Hydrophilic count":            hydrophilic_count_s,
            "Hydrophobic count":            hydrophobic_count_s,
            "High solubility proxy count":  high_sol_proxy_count,
            "Low solubility proxy count":   low_sol_proxy_count,
            "Most hydrophobic":             most_hydrophobic,
            "Most hydrophilic":             most_hydrophilic,
            "Most stable":                  most_stable,
            "Highest solubility proxy":     highest_proxy_solubility
        }

        physchem_summary_df = pd.DataFrame([
            {"Metric": k, "Value": v} for k, v in physchem_summary.items()
        ])

        summary_csv = PHYCHEM_OUTPUT_DIR / "physicochemical_summary.csv"
        physchem_summary_df.to_csv(summary_csv, index=False)
        output_files.append(summary_csv)

        # ============================================================
        # Statistical Summary
        # ============================================================

        stats_summary_df = pd.DataFrame()

        if show_stats_ranking:
            valid_numeric_cols = [
                col for col in numeric_summary_columns
                if col in physchem_df.columns
                and pd.api.types.is_numeric_dtype(physchem_df[col])
            ]
            if valid_numeric_cols:
                stats_summary_df = physchem_df[valid_numeric_cols].describe().T.reset_index()
                stats_summary_df = stats_summary_df.rename(
                    columns={"index": "Feature"}
                ).round(3)
                stats_summary_csv = PHYCHEM_OUTPUT_DIR / "physicochemical_statistical_summary.csv"
                stats_summary_df.to_csv(stats_summary_csv, index=False)
                output_files.append(stats_summary_csv)
            else:
                add_plot_note(plot_notes, "Statistical Summary", "Skipped",
                              "No numeric columns available.")

        # ============================================================
        # Indicator Explanation Table
        # ============================================================

        added_indicators_explanation_df = pd.DataFrame([
            {"Indicator": "Aliphatic Index",
             "Meaning": "Contribution of Ala, Val, Ile, Leu to aliphatic character.",
             "Reference": "Ikai AJ (1980) J Biochem 88(6):1895-1898"},
            {"Indicator": "Extinction Coefficient",
             "Meaning": "Theoretical UV absorbance-related coefficient (Trp, Tyr, Cys).",
             "Reference": "Biopython ProteinAnalysis.molar_extinction_coefficient()"},
            {"Indicator": "Amino Acid Composition",
             "Meaning": "Counts and percentages of the 20 standard amino acids.",
             "Reference": "—"},
            {"Indicator": "Hydrophobic Residues (%)",
             "Meaning": "Percentage of I, V, L, F, M, A, C (KD score > 0).",
             "Reference": "Kyte & Doolittle (1982) J Mol Biol 157:105-132"},
            {"Indicator": "Charged Residues (%)",
             "Meaning": "Percentage of K, R, D, E residues.",
             "Reference": "—"},
            {"Indicator": "Cysteine Count",
             "Meaning": "Number of Cys residues; relevant for disulfide bonding.",
             "Reference": "—"},
            {"Indicator": "GRAVY Class",
             "Meaning": "Hydrophilic (< -0.4), Neutral/mixed, or Hydrophobic (> 0.4).",
             "Reference": "Kyte & Doolittle (1982); DOI:10.3390/ijms24032481"},
            {"Indicator": "Instability Class",
             "Meaning": "Stable (< 40) or Potentially unstable (≥ 40).",
             "Reference": "DOI:10.2174/0929866526666190228144219"},
            {"Indicator": "pI Class",
             "Meaning": "Acidic (< 6), Near-neutral, or Basic (> 8).",
             "Reference": "—"},
            {"Indicator": "Charge Density",
             "Meaning": "Net charge at pH 7 / sequence length. Derived metric.",
             "Reference": "—"},
            {"Indicator": "Solubility Proxy Score",
             "Meaning": (
                 "Heuristic score (0–100%) based on Instability, GRAVY, and Net Charge. "
                 "NOT a validated model."
             ),
             "Reference": "DOI:10.2174/..., DOI:10.3390/..., DOI:10.1038/srep03333"}
        ])

        added_indicators_csv = PHYCHEM_OUTPUT_DIR / "added_indicators_explanation.csv"
        added_indicators_explanation_df.to_csv(added_indicators_csv, index=False)
        output_files.append(added_indicators_csv)

        # ============================================================
        # Optional Plots
        # ============================================================

        plot_sections_html = ""

        if plot_requested:
            plot_numeric_cols = [
                col for col in [
                    "Analyzed Length", "Mol. Weight (kDa)", "pI", "GRAVY",
                    "Net Charge (pH7)", "Instability Index", "Aliphatic Index",
                    "Hydrophobic (%)", "Charged Residues (%)", "Solubility Proxy Score (%)"
                ]
                if col in physchem_df.columns
                and pd.api.types.is_numeric_dtype(physchem_df[col])
            ]

            # ---- Distribution Plot (Box / Violin) ----
            if show_box_violin:
                try:
                    if len(plot_numeric_cols) == 0:
                        add_plot_note(plot_notes, "Distribution Plot", "Skipped",
                                      "No numeric features available.")
                    elif analyzed_count < 2:
                        add_plot_note(plot_notes, "Distribution Plot", "Skipped",
                                      "At least 2 proteins required.")
                    else:
                        selected_cols = plot_numeric_cols[:8]
                        plot_df = physchem_df[["Protein Name"] + selected_cols].copy()
                        normalized_df = plot_df.copy()
                        for col in selected_cols:
                            std_val  = normalized_df[col].std()
                            mean_val = normalized_df[col].mean()
                            normalized_df[col] = (
                                0 if (pd.isna(std_val) or std_val == 0)
                                else (normalized_df[col] - mean_val) / std_val
                            )
                        melted_df = normalized_df.melt(
                            id_vars=["Protein Name"], value_vars=selected_cols,
                            var_name="Feature", value_name="Z-score"
                        )
                        if analyzed_count < MIN_N_FOR_DISTRIBUTION_PLOT:
                            add_plot_caution(
                                plot_cautions, "Distribution Plot",
                                f"Only {analyzed_count} proteins — violin not reliable; "
                                f"box plot shown instead."
                            )
                            fig_dist = px.box(
                                melted_df, x="Feature", y="Z-score", points="all",
                                color="Feature",
                                title=f"Normalized Feature Distribution — Box/Points Plot (n={analyzed_count})"
                            )
                        else:
                            fig_dist = px.violin(
                                melted_df, x="Feature", y="Z-score", box=True,
                                points="all", color="Feature",
                                title=f"Normalized Feature Distribution — Violin + Box + Points Plot (n={analyzed_count})"
                            )
                        apply_publication_text_axis_style(
                            fig_dist,
                            width=1100,
                            height=650,
                            showlegend=False,
                            x_title="Physico-chemical feature",
                            y_title="Z-score normalized value",
                            tickangle=-35,
                            left_margin=90,
                            right_margin=90,
                            top_margin=120,
                            bottom_margin=170,
                        )

                        distribution_html = save_plotly_figure_outputs(
                            fig_dist,
                            base_name="physchem_distribution_plot",
                            output_dir=PHYCHEM_OUTPUT_DIR,
                            output_files=output_files,
                            plot_notes=plot_notes,
                            plot_label="Distribution Plot",
                        )
                        plot_sections_html += f"""
                        <section id="distribution-plot">
                            <h2>Normalized Feature Distribution Plot</h2>
                            <div class="academic-note">
                                Features are z-score normalized. With n={analyzed_count} protein(s),
                                this plot is descriptive only.
                            </div>
                            {fig_dist.to_html(full_html=False, include_plotlyjs=False)}
                        </section>"""
                except Exception as e:
                    add_plot_note(plot_notes, "Distribution Plot", "Failed", str(e))

            # ---- Correlation Heatmap ----
            if show_heatmap:
                try:
                    if len(plot_numeric_cols) < 2:
                        add_plot_note(plot_notes, "Heatmap", "Skipped",
                                      "At least 2 numeric features required.")
                    elif analyzed_count < 2:
                        add_plot_note(plot_notes, "Heatmap", "Skipped",
                                      "At least 2 proteins required.")
                    else:
                        if analyzed_count < MIN_N_FOR_CORRELATION:
                            add_plot_caution(
                                plot_cautions, "Feature Correlation Heatmap",
                                f"Only {analyzed_count} proteins — correlations are "
                                f"mathematically valid but statistically unreliable. "
                                f"Interpret as exploratory only."
                            )

                        # Remove only constant / zero-variance columns from the heatmap.
                        # They produce NaN correlations and appear as blank white rows/columns.
                        # The original result table and all calculations remain unchanged.
                        heatmap_cols = [
                            col for col in plot_numeric_cols
                            if physchem_df[col].nunique(dropna=True) > 1
                        ]
                        removed_constant_cols = [
                            col for col in plot_numeric_cols if col not in heatmap_cols
                        ]
                        if removed_constant_cols:
                            add_plot_caution(
                                plot_cautions, "Feature Correlation Heatmap",
                                "Constant / zero-variance feature(s) removed from heatmap "
                                "to avoid blank NaN rows/columns: "
                                + ", ".join(removed_constant_cols)
                            )

                        if len(heatmap_cols) < 2:
                            add_plot_note(
                                plot_notes, "Heatmap", "Skipped",
                                "At least 2 non-constant numeric features are required."
                            )
                        else:
                            corr = physchem_df[heatmap_cols].corr().round(2)
                            heatmap_size = max(720, min(1050, 72 * len(heatmap_cols) + 260))
                            fig_hm = px.imshow(
                                corr, text_auto=".2f", aspect="auto",
                                color_continuous_scale="RdBu_r", zmin=-1, zmax=1,
                                title=f"Feature Correlation Heatmap (n={analyzed_count})"
                            )
                            apply_publication_text_axis_style(
                                fig_hm,
                                width=heatmap_size + 190,
                                height=heatmap_size,
                                left_margin=205,
                                right_margin=125,
                                top_margin=105,
                                bottom_margin=175,
                                tickangle=-38,
                            )
                            fig_hm.update_traces(xgap=0, ygap=0)
                            fig_hm.update_yaxes(tickangle=0, automargin=True, showgrid=False)
                            fig_hm.update_xaxes(tickangle=-38, automargin=True, showgrid=False)
                            fig_hm.update_coloraxes(
                                colorbar=dict(
                                    tickfont=dict(color=PLOT_TEXT_COLOR, size=12),
                                    title=dict(font=dict(color=PLOT_TEXT_COLOR, size=13)),
                                    x=1.015,
                                    xpad=8,
                                    len=0.78,
                                )
                            )

                            heatmap_html = save_plotly_figure_outputs(
                                fig_hm,
                                base_name="physchem_heatmap",
                                output_dir=PHYCHEM_OUTPUT_DIR,
                                output_files=output_files,
                                plot_notes=plot_notes,
                                plot_label="Heatmap",
                            )
                            plot_sections_html += f"""
                            <section id="correlation-heatmap">
                                <h2>Feature Correlation Heatmap</h2>
                                <div class="academic-note">
                                    Correlation analysis with n={analyzed_count} protein(s).
                                    {'⚠️ Small sample — coefficients are exploratory only.'
                                     if analyzed_count < MIN_N_FOR_CORRELATION
                                     else 'Exploratory; do not over-interpret.'}
                                </div>
                                {fig_hm.to_html(full_html=False, include_plotlyjs=False)}
                            </section>"""
                except Exception as e:
                    add_plot_note(plot_notes, "Heatmap", "Failed", str(e))

            # ---- Parallel Coordinates ----
            if show_parallel:
                try:
                    if len(plot_numeric_cols) < 2:
                        add_plot_note(plot_notes, "Parallel Coordinates", "Skipped",
                                      "At least 2 numeric features required.")
                    elif analyzed_count < 2:
                        add_plot_note(plot_notes, "Parallel Coordinates", "Skipped",
                                      "At least 2 proteins required.")
                    else:
                        if analyzed_count < MIN_N_FOR_PARALLEL:
                            add_plot_caution(
                                plot_cautions, "Parallel Coordinates",
                                f"Only {analyzed_count} proteins — "
                                f"patterns should not be over-interpreted."
                            )
                        parallel_cols = plot_numeric_cols[:8]
                        parallel_df   = physchem_df[["Protein Name"] + parallel_cols].copy()

                        # ✅ FIX: keep the raw color column BEFORE normalization
                        SOL_COL = "Solubility Proxy Score (%)"
                        normalized_parallel_df = parallel_df.copy()
                        for col in parallel_cols:
                            mn = normalized_parallel_df[col].min()
                            mx = normalized_parallel_df[col].max()
                            normalized_parallel_df[col] = (
                                0.5 if (pd.isna(mn) or mn == mx)
                                else (normalized_parallel_df[col] - mn) / (mx - mn)
                            )

                        # ✅ FIX: add un-normalized solubility col for color encoding
                        if SOL_COL in physchem_df.columns and SOL_COL not in parallel_cols:
                            normalized_parallel_df[SOL_COL] = physchem_df[SOL_COL].values

                        color_column = (
                            SOL_COL
                            if SOL_COL in normalized_parallel_df.columns
                            else parallel_cols[-1]
                        )

                        fig_pc = px.parallel_coordinates(
                            normalized_parallel_df[parallel_cols],
                            color=normalized_parallel_df[color_column],
                            title=(
                                f"Parallel Coordinates — Min-Max Normalized Features "
                                f"(n={analyzed_count})"
                            ),
                            color_continuous_scale="Viridis"
                        )
                        apply_publication_text_axis_style(
                            fig_pc,
                            width=1100,
                            height=650,
                            left_margin=80,
                            right_margin=140,
                            top_margin=120,
                            bottom_margin=80,
                        )

                        parallel_html = save_plotly_figure_outputs(
                            fig_pc,
                            base_name="physchem_parallel_coordinates",
                            output_dir=PHYCHEM_OUTPUT_DIR,
                            output_files=output_files,
                            plot_notes=plot_notes,
                            plot_label="Parallel Coordinates",
                        )
                        plot_sections_html += f"""
                        <section id="parallel-coordinates">
                            <h2>Parallel Coordinates</h2>
                            <div class="academic-note">
                                Values are min-max normalized. n={analyzed_count}. Descriptive only.
                            </div>
                            {fig_pc.to_html(full_html=False, include_plotlyjs=False)}
                        </section>"""
                except Exception as e:
                    add_plot_note(plot_notes, "Parallel Coordinates", "Failed", str(e))

            # ---- Scatter Matrix ----
            if show_scatter_matrix:
                try:
                    if len(plot_numeric_cols) < 2:
                        add_plot_note(plot_notes, "Scatter Matrix", "Skipped",
                                      "At least 2 numeric features required.")
                    elif analyzed_count < 2:
                        add_plot_note(plot_notes, "Scatter Matrix", "Skipped",
                                      "At least 2 proteins required.")
                    else:
                        if analyzed_count < MIN_N_FOR_SCATTER:
                            add_plot_caution(
                                plot_cautions, "Scatter Matrix",
                                f"Only {analyzed_count} proteins — "
                                f"scatter patterns are exploratory only."
                            )

                        # Use a custom pairwise scatter matrix instead of px.scatter_matrix.
                        # Reason: px.scatter_matrix can compress vertical labels/ticks unevenly
                        # when one feature has a narrow numeric range, such as GRAVY.
                        # This custom version keeps equal panel spacing and controls tick density.
                        scatter_cols = plot_numeric_cols[:5]
                        scatter_label_map = {
                            "Analyzed Length": "Length",
                            "Mol. Weight (kDa)": "MW (kDa)",
                            "pI": "pI",
                            "GRAVY": "GRAVY",
                            "Net Charge (pH7)": "Net Charge",
                            "Instability Index": "Instability",
                            "Aliphatic Index": "Aliphatic",
                            "Hydrophobic (%)": "Hydrophobic (%)",
                            "Charged Residues (%)": "Charged (%)",
                            "Solubility Proxy Score (%)": "Solubility (%)",
                        }
                        n_scatter = len(scatter_cols)
                        scatter_color_col = (
                            "Solubility Proxy Score (%)"
                            if "Solubility Proxy Score (%)" in physchem_df.columns
                            else None
                        )
                        scatter_color_values = (
                            physchem_df[scatter_color_col]
                            if scatter_color_col is not None
                            else None
                        )

                        fig_sm = make_subplots(
                            rows=n_scatter,
                            cols=n_scatter,
                            horizontal_spacing=0.035,
                            vertical_spacing=0.035,
                        )

                        colorbar_added = False
                        for row_i, y_col in enumerate(scatter_cols, start=1):
                            for col_i, x_col in enumerate(scatter_cols, start=1):
                                if row_i == col_i:
                                    # Diagonal cells are used only as clean variable labels.
                                    fig_sm.add_trace(
                                        go.Scatter(
                                            x=[0.5],
                                            y=[0.5],
                                            mode="text",
                                            text=[scatter_label_map.get(x_col, x_col)],
                                            textfont=dict(
                                                family=PLOT_FONT_FAMILY,
                                                size=14,
                                                color=PLOT_TEXT_COLOR,
                                            ),
                                            hoverinfo="skip",
                                            showlegend=False,
                                        ),
                                        row=row_i,
                                        col=col_i,
                                    )
                                    fig_sm.update_xaxes(
                                        range=[0, 1],
                                        showticklabels=False,
                                        showgrid=False,
                                        zeroline=False,
                                        showline=False,
                                        row=row_i,
                                        col=col_i,
                                    )
                                    fig_sm.update_yaxes(
                                        range=[0, 1],
                                        showticklabels=False,
                                        showgrid=False,
                                        zeroline=False,
                                        showline=False,
                                        row=row_i,
                                        col=col_i,
                                    )
                                    continue

                                marker_style = dict(size=8, opacity=0.78)
                                if scatter_color_values is not None:
                                    marker_style.update(
                                        color=scatter_color_values,
                                        colorscale="Viridis",
                                        showscale=not colorbar_added,
                                        colorbar=(
                                            dict(
                                                title=dict(
                                                    text="Solubility<br>Proxy (%)",
                                                    font=dict(color=PLOT_TEXT_COLOR, size=13),
                                                    side="right",
                                                ),
                                                tickfont=dict(color=PLOT_TEXT_COLOR, size=12),
                                                x=1.02,
                                                xpad=8,
                                                len=0.78,
                                            )
                                            if not colorbar_added
                                            else None
                                        ),
                                    )
                                    colorbar_added = True

                                fig_sm.add_trace(
                                    go.Scatter(
                                        x=physchem_df[x_col],
                                        y=physchem_df[y_col],
                                        mode="markers",
                                        marker=marker_style,
                                        showlegend=False,
                                        hovertemplate=(
                                            "Protein: %{customdata}<br>"
                                            + scatter_label_map.get(x_col, x_col)
                                            + ": %{x}<br>"
                                            + scatter_label_map.get(y_col, y_col)
                                            + ": %{y}<extra></extra>"
                                        ),
                                        customdata=physchem_df["Protein Name"],
                                    ),
                                    row=row_i,
                                    col=col_i,
                                )

                        fig_sm.update_layout(
                            template="plotly_white",
                            width=1250,
                            height=max(1050, 225 * n_scatter),
                            paper_bgcolor=PLOT_BACKGROUND_COLOR,
                            plot_bgcolor=PLOT_BACKGROUND_COLOR,
                            title=dict(
                                text=f"Scatter Matrix — Selected Features (n={analyzed_count})",
                                x=0.5,
                                xanchor="center",
                                y=0.965,
                                yanchor="top",
                                font=dict(size=22, color=PLOT_TEXT_COLOR),
                                pad=dict(t=8, b=22),
                            ),
                            font=dict(
                                family=PLOT_FONT_FAMILY,
                                color=PLOT_TEXT_COLOR,
                                size=13,
                            ),
                            margin=dict(l=145, r=155, t=120, b=135),
                            showlegend=False,
                        )

                        # Equal spacing for all panels; keep tick density low so GRAVY
                        # does not look closer or more compressed than other rows.
                        fig_sm.update_xaxes(
                            showline=False,
                            zeroline=False,
                            showgrid=True,
                            gridcolor="rgba(0,0,0,0.08)",
                            tickfont=dict(color=PLOT_TEXT_COLOR, size=11),
                            title_font=dict(color=PLOT_TEXT_COLOR, size=13),
                            nticks=3,
                            automargin=True,
                            title_standoff=16,
                        )
                        fig_sm.update_yaxes(
                            showline=False,
                            zeroline=False,
                            showgrid=True,
                            gridcolor="rgba(0,0,0,0.08)",
                            tickfont=dict(color=PLOT_TEXT_COLOR, size=11),
                            title_font=dict(color=PLOT_TEXT_COLOR, size=13),
                            nticks=3,
                            automargin=True,
                            title_standoff=28,
                        )

                        # Only outer axes receive titles/tick labels. This prevents the
                        # vertical axis labels from stacking unevenly inside the figure.
                        for row_i, y_col in enumerate(scatter_cols, start=1):
                            for col_i, x_col in enumerate(scatter_cols, start=1):
                                if row_i != n_scatter:
                                    fig_sm.update_xaxes(showticklabels=False, row=row_i, col=col_i)
                                if col_i != 1:
                                    fig_sm.update_yaxes(showticklabels=False, row=row_i, col=col_i)

                        for col_i, x_col in enumerate(scatter_cols, start=1):
                            fig_sm.update_xaxes(
                                title_text=scatter_label_map.get(x_col, x_col),
                                row=n_scatter,
                                col=col_i,
                            )
                        for row_i, y_col in enumerate(scatter_cols, start=1):
                            fig_sm.update_yaxes(
                                title_text=scatter_label_map.get(y_col, y_col),
                                row=row_i,
                                col=1,
                            )

                        scatter_html = save_plotly_figure_outputs(
                            fig_sm,
                            base_name="physchem_scatter_matrix",
                            output_dir=PHYCHEM_OUTPUT_DIR,
                            output_files=output_files,
                            plot_notes=plot_notes,
                            plot_label="Scatter Matrix",
                        )
                        plot_sections_html += f"""
                        <section id="scatter-matrix">
                            <h2>Scatter Matrix</h2>
                            <div class="academic-note">
                                n={analyzed_count}.
                                {'⚠️ Small sample — exploratory only.'
                                 if analyzed_count < MIN_N_FOR_SCATTER
                                 else 'Exploratory with small protein sets.'}
                            </div>
                            {fig_sm.to_html(full_html=False, include_plotlyjs=False)}
                        </section>"""
                except Exception as e:
                    add_plot_note(plot_notes, "Scatter Matrix", "Failed", str(e))

        plot_caution_df = pd.DataFrame(plot_cautions)
        plot_notes_df   = pd.DataFrame(plot_notes)
        plot_caution_df.to_csv(plot_caution_csv, index=False)
        plot_notes_df.to_csv(plot_notes_csv, index=False)
        output_files.extend([plot_caution_csv, plot_notes_csv])

        # ============================================================
        # HTML Report
        # ============================================================

        if export_html_report:
            plotly_script = (
                '<script src="https://cdn.plot.ly/plotly-latest.min.js"></script>'
                if "plotly-graph-div" in plot_sections_html else ""
            )

            plot_nav_links = ""
            if show_box_violin:     plot_nav_links += '<a href="#distribution-plot">Distribution Plot</a>'
            if show_heatmap:        plot_nav_links += '<a href="#correlation-heatmap">Correlation Heatmap</a>'
            if show_parallel:       plot_nav_links += '<a href="#parallel-coordinates">Parallel Coordinates</a>'
            if show_scatter_matrix: plot_nav_links += '<a href="#scatter-matrix">Scatter Matrix</a>'

            stats_table_html = (
                df_to_html_report_table(stats_summary_df)
                if not stats_summary_df.empty
                else "<p>Statistical summary was not generated.</p>"
            )

            html_report = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>FoldColabFlow Physico-Chemical Analysis Report</title>
{plotly_script}
{REPORT_CSS}
</head>
<body>

<header>
    <h1>🧪 FoldColabFlow Physico-Chemical Analysis Report</h1>
    <p>{analyzed_count} protein(s) analyzed | {skipped_count} skipped | {warning_count} warning(s)</p>
</header>

<div class="layout">
<aside class="sidebar">
    <h3>Navigation</h3>
    <a href="#summary">Summary</a>
    <a href="#method">Solubility Proxy Method</a>
    <a href="#core">Core Properties</a>
    <a href="#additional">Additional Indicators</a>
    <a href="#stats">Statistical Summary</a>
    <a href="#indicator-explanation">Indicator Explanation</a>
    <a href="#warnings">Warnings</a>
    <a href="#skipped">Skipped Sequences</a>
    <a href="#plot-cautions">Plot Cautions</a>
    <a href="#plot-notes">Plot Notes</a>
    {plot_nav_links}
</aside>

<main class="content">

<section id="summary">
    <h2>Summary</h2>
    {df_to_html_report_table(physchem_summary_df)}
</section>

<section id="method">
    <h2>Solubility Proxy Score Method</h2>
    <div class="note">
        <b>Important:</b> Solubility Proxy Score is a
        <b>heuristic screening score, not a validated solubility prediction model</b>.<br><br>
        <b>Criteria:</b><br>
        +40 pts if Instability Index &lt; 40 — Ref: DOI:10.2174/0929866526666190228144219<br>
        +40 pts if GRAVY &lt; -0.4 — Ref: DOI:10.3390/ijms24032481<br>
        +20 pts if |Net charge at pH7| &ge; 4.5 — Ref: DOI:10.1038/srep03333<br><br>
        Score = (raw_pts / 100) &times; 100 (%)<br><br>
        For validated solubility prediction: use <b>Protein-Sol, SoluProt, or CamSol</b>.
    </div>
</section>

<section id="core">
    <h2>Core Physico-Chemical Properties</h2>
    <div class="academic-note">
        Main interpretable indicators for initial sequence-based screening.
    </div>
    {df_to_html_report_table(core_preview_df)}
</section>

<section id="additional">
    <h2>Additional Indicators</h2>
    <div class="academic-note">
        Deeper interpretation indicators separated for readability.
    </div>
    {df_to_html_report_table(additional_preview_df)}
</section>

<section id="stats">
    <h2>Statistical Summary</h2>
    <div class="academic-note">
        Descriptive only. Do not over-interpret with small protein sets.
    </div>
    {stats_table_html}
</section>

<section id="indicator-explanation">
    <h2>Indicator Explanation &amp; References</h2>
    {df_to_html_report_table(added_indicators_explanation_df)}
</section>

<section id="warnings">
    <h2>Warnings</h2>
    <div class="warning">
        These warnings are theoretical sequence-based flags. Interpret cautiously.
    </div><br>
    {df_to_html_report_table(physchem_warning_df)
     if not physchem_warning_df.empty else "<p>No warnings.</p>"}
</section>

<section id="skipped">
    <h2>Skipped Sequences</h2>
    {df_to_html_report_table(physchem_skipped_df)
     if not physchem_skipped_df.empty else "<p>No skipped sequences.</p>"}
</section>

<section id="plot-cautions">
    <h2>Plot Cautions</h2>
    <div class="academic-note">
        Notes added when protein count is below recommended thresholds.
    </div>
    {df_to_html_report_table(plot_caution_df)
     if not plot_caution_df.empty else "<p>No plot cautions.</p>"}
</section>

<section id="plot-notes">
    <h2>Plot Generation Notes</h2>
    {df_to_html_report_table(plot_notes_df)
     if not plot_notes_df.empty else "<p>No plot notes.</p>"}
</section>

{plot_sections_html}

</main>
</div>
</body>
</html>"""

            with open(html_report_path, "w", encoding="utf-8") as f:
                f.write(html_report)
            output_files.append(html_report_path)

        # ============================================================
        # ZIP
        # ============================================================

        zip_path = PHYCHEM_OUTPUT_DIR / "physicochemical_outputs.zip"
        create_zip_from_files(output_files, zip_path)

        # ============================================================

        # ============================================================
        # Colab Display
        # ============================================================

        core_table_preview_html = highlight_min_max_table_html(
            core_preview_df, max_rows=TABLE_PREVIEW_ROWS
        )

        if physchem_warning_df.empty and physchem_skipped_df.empty:
            warning_line_html = (
                '<div class="fi-box"><h4>✅ Warning Files</h4>'
                '<p style="margin-bottom:0;color:#166534;">'
                'No warnings or skipped sequences detected.</p></div>'
            )
        else:
            warning_line_html = (
                '<div class="fi-warning-box"><h4>🚨 Warning Files</h4>'
                '<p style="color:#7f1d1d;margin-bottom:0;">'
                '<b>' + str(warning_count) + '</b> warning flag(s), '
                '<b>' + str(skipped_count) + '</b> skipped sequence(s).'
                ' See HTML report and ZIP for details.</p></div>'
            )

        if show_stats_ranking:
            if stats_summary_df.empty:
                stats_section_html = (
                    '<div class="fi-box"><h3>📈 Statistical Summary</h3>'
                    '<p class="fi-small">No numeric columns available.</p></div>'
                )
            else:
                stats_section_html = (
                    '<div class="fi-box"><h3>📈 Statistical Summary</h3>'
                    + preview_table_html(stats_summary_df, max_rows=20)
                    + '</div>'
                )
        else:
            stats_section_html = ""

        if show_table:
            table_section_html = (
                '<div class="fi-box">'
                '<h3>📋 Core Physico-Chemical Table Preview</h3>'
                '<p class="fi-small">Min/max highlighted in preview. Full tables in HTML report &amp; ZIP.</p>'
                + core_table_preview_html
                + '</div>'
            )
        else:
            table_section_html = ""

        plot_names = []
        if show_box_violin:     plot_names.append("Distribution Plot")
        if show_heatmap:        plot_names.append("Heatmap")
        if show_parallel:       plot_names.append("Parallel Coordinates")
        if show_scatter_matrix: plot_names.append("Scatter Matrix")
        plot_line = ", ".join(plot_names)

        caution_note = ""
        if not plot_caution_df.empty:
            caution_note = "<br>⚠️ Small sample size — see Plot Cautions section."

        if plot_line:
            plot_message_html = (
                '<div class="fi-box"><h4>📊 Optional Plots</h4>'
                '<p class="fi-small" style="margin-bottom:0;">'
                'Generated in Colab and saved to HTML/ZIP plus high-resolution PNG/SVG: <b>'
                + escape(plot_line) + '</b>.' + caution_note
                + '</p></div>'
            )
        else:
            plot_message_html = (
                '<div class="fi-box"><h4>📊 Optional Plots</h4>'
                '<p class="fi-small" style="margin-bottom:0;">Plot generation disabled.</p></div>'
            )

        output_links_html = make_download_link(zip_path, "Download all outputs as ZIP")
        if export_html_report and html_report_path.exists():
            output_links_html += make_download_link(html_report_path, "Download HTML Report")

        avg_mw   = round(physchem_df["Mol. Weight (kDa)"].mean(), 2)
        avg_pi   = round(physchem_df["pI"].mean(), 2)
        avg_grav = round(physchem_df["GRAVY"].mean(), 3)

        colab_html = (
            STYLE_BLOCK
            + '<div class="fi-container">'
            + '<h2 class="fi-title">🧪 Physico-Chemical Properties Analysis</h2>'

            + '<div class="fi-box">'
            + '<h3>📍 Current Stage: Physico-Chemical Characterization</h3>'
            + '<p style="line-height:1.8;margin-bottom:0;">'
            + 'Sequence-based theoretical properties via ProtParam-style descriptors. '
            + 'Includes a heuristic <b>Solubility Proxy Score</b> for initial screening '
            + '(not a validated model).</p></div>'

            + '<div class="fi-box"><h3>📊 Compact Results</h3>'
            + '<div class="fi-stat-grid">'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Analyzed</div>'
            + '<div class="fi-stat-value">' + str(analyzed_count) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Skipped</div>'
            + '<div class="fi-stat-value">' + str(skipped_count) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Warnings</div>'
            + '<div class="fi-stat-value">' + str(warning_count) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Avg. MW</div>'
            + '<div class="fi-stat-value">' + str(avg_mw) + ' kDa</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Avg. pI</div>'
            + '<div class="fi-stat-value">' + str(avg_pi) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Avg. GRAVY</div>'
            + '<div class="fi-stat-value">' + str(avg_grav) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Stable proteins</div>'
            + '<div class="fi-stat-value">' + str(stable_count) + '</div></div>'
            + '<div class="fi-stat-card"><div class="fi-stat-label">Low sol. proxy</div>'
            + '<div class="fi-stat-value">' + str(low_sol_proxy_count) + '</div></div>'
            + '</div>'
            + '<p style="line-height:1.8;margin-top:14px;margin-bottom:0;">'
            + '<b>Most hydrophobic:</b> ' + escape(str(most_hydrophobic)) + '<br>'
            + '<b>Most hydrophilic:</b> ' + escape(str(most_hydrophilic)) + '<br>'
            + '<b>Most stable:</b> ' + escape(str(most_stable)) + '<br>'
            + '<b>Highest solubility proxy:</b> ' + escape(str(highest_proxy_solubility))
            + '</p></div>'

            + '<div class="fi-box"><h3>🧪 Solubility Proxy Score Method</h3>'
            + '<div class="fi-method">'
            + '<b>Heuristic screening score — NOT a validated solubility prediction model.</b><br>'
            + '+40 pts: Instability Index &lt; 40 — Ref: DOI:10.2174/0929866526666190228144219<br>'
            + '+40 pts: GRAVY &lt; -0.4 — Ref: DOI:10.3390/ijms24032481<br>'
            + '+20 pts: |Net Charge at pH7| &ge; 4.5 — Ref: DOI:10.1038/srep03333<br>'
            + 'Score normalized to 100%.<br>'
            + '<b>Hydrophobic residues</b> defined as KD score &gt; 0: I, V, L, F, M, A, C — '
            + 'Ref: Kyte &amp; Doolittle (1982).'
            + '</div></div>'

            + warning_line_html
            + plot_message_html
            + stats_section_html
            + table_section_html

            + '<div class="fi-box"><h3>📁 Output Files</h3>'
            + '<p style="line-height:1.8;margin-bottom:10px;">Saved in: <code>'
            + escape(str(PHYCHEM_OUTPUT_DIR)) + '</code></p>'
            + output_links_html
            + '</div>'

            + '<div class="fi-next">'
            + '<p style="margin:0;font-size:16px;">'
            + '➡️ <b>Next stage:</b> 3D Structure Prediction with ColabFold'
            + '</p></div>'

            + '</div>'
            + RESIZE_SCRIPT
        )

        display(HTML(colab_html))

        # ============================================================
        # Save Globals
        # ============================================================

        globals().update({
            "physchem_df":                     physchem_df,
            "physchem_core_df":                core_preview_df,
            "physchem_additional_df":          additional_preview_df,
            "physchem_summary":                physchem_summary,
            "physchem_summary_df":             physchem_summary_df,
            "physchem_stats_summary_df":       stats_summary_df,
            "physchem_warning_df":             physchem_warning_df,
            "physchem_skipped_df":             physchem_skipped_df,
            "physchem_plot_notes_df":          plot_notes_df,
            "physchem_plot_caution_df":        plot_caution_df,
            "aa_count_df":                     aa_count_df,
            "aa_percent_df":                   aa_percent_df,
            "added_indicators_explanation_df": added_indicators_explanation_df,
            "physchem_output_dir":             str(PHYCHEM_OUTPUT_DIR),
            "physchem_zip_file":               str(zip_path),
            "physchem_report_file":            (
                str(html_report_path) if export_html_report else None
            )
        })

        if "save_state" in globals():
            save_state(
                "03_physicochemical_analysis",
                physchem_df=physchem_df,
                physchem_core_df=core_preview_df,
                physchem_additional_df=additional_preview_df,
                physchem_summary=physchem_summary,
                physchem_summary_df=physchem_summary_df,
                physchem_stats_summary_df=stats_summary_df,
                physchem_warning_df=physchem_warning_df,
                physchem_skipped_df=physchem_skipped_df,
                physchem_plot_notes_df=plot_notes_df,
                physchem_plot_caution_df=plot_caution_df,
                aa_count_df=aa_count_df,
                aa_percent_df=aa_percent_df,
                added_indicators_explanation_df=added_indicators_explanation_df,
                physchem_output_dir=str(PHYCHEM_OUTPUT_DIR),
                physchem_zip_file=str(zip_path),
                physchem_report_file=(
                    str(html_report_path) if export_html_report else None
                )
            )
```

```python
#@title 🧬 Functional Annotation (PSALM CLI Stable - Clean Version)

!pip -q install biopython pandas protein-sequence-annotation==2.1.12

import os
import glob
import tempfile
import subprocess
import pandas as pd

from Bio.PDB import PDBParser
from Bio.SeqUtils import seq1
from IPython.display import display, HTML

print("🧬 ================= PSALM FUNCTIONAL ANNOTATION =================\n")

# ==========================================
# JOB
# ==========================================

if "jobname" not in globals():
    jobname = input("Enter jobname: ")

folders = sorted([
    x for x in glob.glob(f"{jobname}/Seq_*")
    if os.path.isdir(x)
])

if len(folders) == 0:
    print("❌ No sequence folders found.")
    raise SystemExit()

print("\n📂 Available sequences:\n")
for i, f in enumerate(folders, start=1):
    print(f"{i}. {os.path.basename(f)}")

idx = int(input("\n👉 Select sequence index: ") or 1) - 1
idx = max(0, min(idx, len(folders) - 1))

folder = folders[idx]

print(f"\n🔄 Selected: {os.path.basename(folder)}")

# ==========================================
# PDB SELECTION
# ==========================================

pdbs = sorted(glob.glob(f"{folder}/**/*.pdb", recursive=True))

if len(pdbs) == 0:
    print("❌ No PDB files found.")
    raise SystemExit()

print("\n📄 Available structures:\n")
for i, p in enumerate(pdbs, start=1):
    print(f"{i}. {os.path.basename(p)}")

pdb_idx = int(input("\n👉 Select PDB index: ") or 1) - 1
pdb_idx = max(0, min(pdb_idx, len(pdbs) - 1))

pdb = pdbs[pdb_idx]

print(f"\n🔬 Using structure: {os.path.basename(pdb)}")

# ==========================================
# SEQUENCE EXTRACTION
# ==========================================

parser = PDBParser(QUIET=True)
structure = parser.get_structure("p", pdb)

seq = []

for model in structure:
    for chain in model:
        for residue in chain:
            if residue.id[0] == " ":
                try:
                    seq.append(seq1(residue.resname))
                except:
                    pass

sequence = "".join(seq)

print(f"\n📏 Sequence length: {len(sequence)} aa")

# ==========================================
# FASTA FILE
# ==========================================

fa = tempfile.NamedTemporaryFile(suffix=".fa", delete=False)

with open(fa.name, "w") as f:
    f.write(">query\n")
    f.write(sequence)

out = "psalm_hits.tsv"

# ==========================================
# RUN PSALM
# ==========================================

cmd = [
    "psalm-scan",
    "-f", fa.name,
    "--to-tsv", out
]

print("\n🚀 Running PSALM scan...\n")

result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode != 0:
    print("❌ PSALM failed:\n")
    print(result.stderr)
    raise SystemExit()

# ==========================================
# LOAD RESULTS
# ==========================================

if os.path.exists(out):
    df = pd.read_csv(out, sep="\t")
else:
    df = pd.DataFrame()

# ==========================================
# UI OUTPUT (styled like your pipeline)
# ==========================================

display(HTML(f"""
<div style="
    background:#e5e7eb;
    border:2px solid #9ca3af;
    border-radius:16px;
    padding:20px;
    font-family:Arial;
    margin-top:15px;
">

    <h2 style="color:#1d4ed8; text-align:center;">
        🧬 Functional Annotation (PSALM)
    </h2>

    <div style="
        background:#f8fafc;
        border:1px solid #cbd5e1;
        border-radius:12px;
        padding:12px;
        margin-bottom:12px;
    ">
        📁 Folder: {os.path.basename(folder)}<br>
        🔬 Structure: {os.path.basename(pdb)}<br>
        📏 Sequence length: {len(sequence)} aa<br>
        📊 Hits found: {len(df)}
    </div>

</div>
"""))

# ==========================================
# DISPLAY TABLE
# ==========================================

if len(df) > 0:
    display(df)
else:
    print("⚠️ No functional hits found.")

# save globally for pipeline
globals()["functional_df"] = df

print("\n✅ Done")
```

```python
#@markdown ## 🧬 📏 TM-score Structural Alignment

!wget -q https://zhanggroup.org/TM-align/TMalign.cpp
!g++ -O3 -ffast-math -o TMalign TMalign.cpp

import os
import glob
import subprocess
import pandas as pd
import re
from IPython.display import display, HTML

print("🧬 ================= STRUCTURAL ALIGNMENT (TM-align) =================\n")

# =========================
# JOBNAME
# =========================
if "jobname" not in globals():
    jobname = input("📌 Enter project folder name: ").strip()

base_path = jobname

# =========================
# FIND PDB FILES
# =========================
pdb_files = sorted(glob.glob(f"{base_path}/**/*.pdb", recursive=True))

pdb_files = [
    p for p in pdb_files
    if "rank" in os.path.basename(p) or "unrelaxed" in os.path.basename(p)
]

if len(pdb_files) == 0:
    print("❌ No PDB files found!")
    raise SystemExit()

print("\n📂 Available structures:\n")
for i, f in enumerate(pdb_files, start=1):
    print(f"{i}. {os.path.basename(f)}")

# =========================
# USER INPUT
# =========================
idx1 = int(input("\n👉 Select structure 1: ") or 1) - 1
idx2 = int(input("👉 Select structure 2: ") or 2) - 1

idx1 = max(0, min(idx1, len(pdb_files) - 1))
idx2 = max(0, min(idx2, len(pdb_files) - 1))

pdb1 = pdb_files[idx1]
pdb2 = pdb_files[idx2]

# =========================
# RUN TM-align
# =========================
print("\n⚙️ Running TM-align...\n")

cmd = ["./TMalign", pdb1, pdb2]
result = subprocess.run(cmd, capture_output=True, text=True)
output = result.stdout

# =========================
# PARSE OUTPUT
# =========================
tm_scores = re.findall(r"TM-score=\s*([0-9.]+)", output)
rmsd_match = re.search(r"RMSD=\s*([0-9.]+)", output)

tm1 = float(tm_scores[0]) if len(tm_scores) > 0 else None
tm2 = float(tm_scores[1]) if len(tm_scores) > 1 else None
rmsd = float(rmsd_match.group(1)) if rmsd_match else None

if tm1 is None:
    print("❌ TM-align failed or output not parsed correctly")
    raise SystemExit()

# =========================
# SAVE OBJECTS
# =========================
tm_result = {
    "structure1": os.path.basename(pdb1),
    "structure2": os.path.basename(pdb2),
    "tm_score_chain1": tm1,
    "tm_score_chain2": tm2,
    "rmsd": rmsd,
    "raw_output": output
}

tm_df = pd.DataFrame([{
    "Structure 1": tm_result["structure1"],
    "Structure 2": tm_result["structure2"],
    "TM-score (chain1)": tm1,
    "TM-score (chain2)": tm2,
    "RMSD (Å)": rmsd
}])

# =========================
# UI OUTPUT (PIPELINE STYLE)
# =========================
display(HTML(f"""
<div style="
    background:#e5e7eb;
    border:2px solid #9ca3af;
    border-radius:16px;
    padding:20px;
    font-family:Arial;
    margin-top:15px;
">

    <h2 style="color:#1d4ed8; text-align:center;">
        📏 TM-align Structural Comparison
    </h2>

    <div style="
        background:#f8fafc;
        border:1px solid #cbd5e1;
        border-radius:12px;
        padding:12px;
        margin-bottom:12px;
    ">
        🔬 Structure 1: {os.path.basename(pdb1)}<br>
        🔬 Structure 2: {os.path.basename(pdb2)}<br>
        📏 RMSD: {rmsd:.3f} Å<br>
        ⭐ TM-score 1: {tm1:.4f}<br>
        ⭐ TM-score 2: {tm2:.4f}
    </div>



</div>
"""))

# =========================
# TABLE OUTPUT
# =========================
display(tm_df)

print("\n✅ TM-align completed and saved:")
print("➡️ tm_result (dict)")
print("➡️ tm_df (DataFrame)")
```
